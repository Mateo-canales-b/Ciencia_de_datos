## Paso 8: Detección de Duplicados

```python
df.duplicated().sum()
df = df.drop_duplicates()
```

## Paso 9: Limpieza de datos

```python
# Quitar espacios y convertir texto
df['columna'] = df['columna'].str.strip().str.lower()
# Convertir fechas
df['fecha'] = pd.to_datetime(df['fecha'])
```

## Paso 10: Análisis de Variables Categóricas

```python
df['categoria'].value_counts()
sns.countplot(x='categoria', data=df)
plt.xticks(rotation=45)
plt.show()
```

## Paso 11: Detección de Outliers

```python
sns.boxplot(x=df['columna'])
plt.show()
```

```python
# Método IQR
Q1 = df['columna'].quantile(0.25)
Q3 = df['columna'].quantile(0.75)
IQR = Q3 - Q1
filtro = (df['columna'] >= Q1 - 1.5 * IQR) & (df['columna'] <= Q3 + 1.5 * IQR)
df_filtrado = df[filtro]
```

## Paso 12: Agrupaciones por Categoría

```python
df.groupby('categoria')['columna'].mean()
```

## Paso 13: Codificación de Variables Categóricas

```python
# One-hot encoding
df_encoded = pd.get_dummies(df, columns=['columna_categorica'])

# Label encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['columna_categorica'] = le.fit_transform(df['columna_categorica'])
```

## Paso 14: Escalamiento de Variables Numéricas

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# StandardScaler
scaler = StandardScaler()
df[['col1', 'col2']] = scaler.fit_transform(df[['col1', 'col2']])

# MinMaxScaler
minmax = MinMaxScaler()
df[['col1', 'col2']] = minmax.fit_transform(df[['col1', 'col2']])
```

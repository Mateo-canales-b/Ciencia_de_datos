### 1. Definir variables básicas y tipos de datos (1 punto)
- Crea una **lista** que contenga al menos **cinco libros**, donde cada libro sea un **diccionario** con los atributos:

  - `autor` (cadena de caracteres)
  - `precio` (decimal)
  - `stock` (entero)



```python
# 1) Inventario base: lista de diccionarios

inventario= [
    {"titulo": "El nombre del viento", "autor": "Patrick Rothfuss", "precio": 15990, "stock": 3},
    {"titulo": "Cien años de soledad", "autor": "Gabriel García Márquez", "precio": 12990, "stock": 2},
    {"titulo": "La sombra del viento", "autor": "Carlos Ruiz Zafón", "precio": 11990, "stock": 0},
    {"titulo": "Ficciones", "autor": "Jorge Luis Borges", "precio": 8990, "stock": 3},
    {"titulo": "Rayuela", "autor": "Julio Cortázar", "precio": 10990, "stock": 1}
]

```


### 2. Control de flujo (1 punto)
- Implementa una función llamada `mostrar_libros_disponibles()` que recorra la lista de
libros y **muestre** los libros que tienen **más de una unidad en stock** usando un `for` y una condición `if`.


```python
# 2) Mostrar libros disponibles 

def mostrar_libros_disponibles():
    print("Libros con más de 1 unidad en stock:")
    disponibles = False
    for libro in inventario:
        if libro["stock"] > 1:
            disponibles = True
            print(f"- {libro['titulo']} | Autor: {libro['autor']} | Precio: ${libro['precio']:.0f} | Stock: {libro['stock']}")
    if not disponibles:
        print("No hay libros en stock.")
      
```

```python
# Ejecuciión de la funcion
mostrar_libros_disponibles()
```

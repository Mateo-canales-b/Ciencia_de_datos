```python
# 2) Mostrar libros disponibles 

def mostrar_libros_disponibles():
    print("Libros en stock:")
    disponibles = False
    for libro in inventario:
        if libro["stock"] >= 1:
            disponibles = True
            print(f"- {libro['titulo']} | Autor: {libro['autor']} | Precio: ${libro['precio']:.0f} | Stock: {libro['stock']}")
    if not disponibles:
        print("No hay libros en stock.")
      
```

## 4. Función personalizada para simular una compra (2 puntos):
- Crea una función comprar_libros(título, cantidad) que reciba como parámetros el
título del libro y la cantidad a comprar. La función debe:
    - Verificar si el libro está en el inventario y si la cantidad deseada está
disponible.
    - Si la compra es válida, restar la cantidad comprada al stock y mostrar el
monto total de la compra.
    - Si la cantidad solicitada es mayor al stock disponible, mostrar un mensaje
de error.


```python
# 4) funcion de compra
carrito = []
def vaciar_carrito():
    carrito=[]

def comprar_libros(titulo: str, cantidad: int):
    # Buscar si tenemos el libro en el inventario
    libro = next((l for l in inventario if l["titulo"].lower() == titulo.lower()), None)

    # Validaciones de cantidad
    if cantidad == 0:
        print("¿Me está diciendo que quiere comprar 0 libros?")
        return
    elif cantidad < 0:
        print(f"jajaja, Difícil comprar {cantidad} libros.")
        return

    # Validación de existencia
    if libro is None:
        print(f"No tenemos el libro '{titulo}' en nuestro catálogo.")
        return

    # Validación de stock
    if libro['stock'] == 0:
        print(f"El libro '{titulo}' está en el catálogo, pero no hay stock disponible.")
        return
    elif libro['stock'] < cantidad:
        print(f"El libro '{titulo}' está en catálogo, pero no podemos satisfacer su pedido. "
              f"Solo nos quedan {libro['stock']} ejemplares.")
        return

    # Casos particulares
    if cantidad == 1 and libro['stock'] == 1:
        print(f"Genial, justo tenemos el último ejemplar de '{titulo}'.")
    elif cantidad == libro['stock']:
        print(f"Te llevas todas las unidades disponibles de '{titulo}' ({cantidad} en total).")
    elif cantidad < libro['stock']:
        print(f"Tenemos stock suficiente de '{titulo}'. "
              f"Pedido: {cantidad} | Stock disponible: {libro['stock']}")
    
    # Creacion del carrito

    while True:
        decision = input("¿Desea añadirlo al carrito? (Si/No): ").strip().lower()
        if decision == "si":
            libro["stock"] -= cantidad
            
            # Calcular precios 
            precio_unit = float(libro["precio"])
            subtotal = precio_unit * cantidad
            iva= subtotal*0.19
            total_item = subtotal + iva

            carrito.append({
                "titulo": libro["titulo"],
                "cantidad": cantidad,
                "precio_unit": precio_unit,
                "subtotal": total_item
            })
            
            print(f"Añadido al carrito: '{libro['titulo']}' x{cantidad} | "
                  f"Total ítem: ${total_item:,.0f}".replace(",", ".")
                  )
            return True

        elif decision == "no":
            print("No se añadió nada al carrito.")
            return False

        else:
            print("No logré entender su respuesta. Por favor escriba 'Si' o 'No'.")
```

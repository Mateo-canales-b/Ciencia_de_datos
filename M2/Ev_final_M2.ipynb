
### 2. Control de flujo (1 punto)
- Implementa una función llamada `mostrar_libros_disponibles()` que recorra la lista de
libros y **muestre** los libros que tienen **más de una unidad en stock** usando un `for` y una condición `if`.


```python
# 2) Mostrar libros disponibles 

def mostrar_libros_disponibles():
    print("Libros en stock:")
    disponibles = False
    for libro in inventario:
        if libro["stock"] >= 1:
            disponibles = True
            print(f"- {libro['titulo']} | Autor: {libro['autor']} | Precio: ${libro['precio']:.2f} | Stock: {libro['stock']}")
    if not disponibles:
        print("No hay libros en stock.")
      
```

### 3. Condiciones y operadores (1 punto)
- Solicita al usuario que ingrese un **rango de precios (mínimo y máximo)** y utiliza una
sentencia `if / elif / else` para **filtrar** los libros en el rango ingresado y mostrarlos.




```python
# 3) Filtro de precios

def filtro_por_precio(max=1000,min=0):
    max=float(input("Ingrese el precio máximo: "))
    min=float(input("Ingrese el precio mínimo: "))
    encontrados = False
    if min < 0 or max <= 0 :
        print("Rango de precios inválido. Se asume precio "
              "mínimo 0 y máximo 1.000") 
        min=0
        max=1000

    elif min > max:
        print("Rango de precios inválido. Se asume precio "
              "mínimo 0 y máximo 1.000") 
        min=0
        max=1000

    else:
        print(f"Buscando libros entre ${min} y ${max}:")
    count = 0
    for libro in inventario:
        if min <= libro["precio"] <= max:
            print(f"- {libro['titulo']} | ${libro['precio']:.2f} | Stock: {libro['stock']}")
            count += 1
    if count == 0:
        print("No se encontraron libros en ese rango de precios.")
        
```

## 4. Función personalizada para simular una compra (2 puntos):
- Crea una función comprar_libros(título, cantidad) que reciba como parámetros el
título del libro y la cantidad a comprar. La función debe:
    - Verificar si el libro está en el inventario y si la cantidad deseada está
disponible.
    - Si la compra es válida, restar la cantidad comprada al stock y mostrar el
monto total de la compra.
    - Si la cantidad solicitada es mayor al stock disponible, mostrar un mensaje
de error.


```python
# 4) funcion de compra
carrito = []
def vaciar_carrito():
    global carrito
    carrito.clear()
    
def comprar_libros(titulo: str, cantidad: int):
    # Libro
    libro = next((l for l in inventario if l["titulo"].lower() == titulo.strip().lower()), None)
    if not libro:
        print("No encontramos ese título en el inventario.")
        return False

    # Cantidad
    if cantidad == 0:
        print("¿Me está diciendo que quiere comprar 0 libros?, no hay chance!")
        return False
    if cantidad < 0:
        print(f"jajaja, Difícil comprar {cantidad} libros.")
        return False
    if libro["stock"] <= 0:
        print("No hay stock disponible de este título.")
        return False
    if cantidad > libro["stock"]:
        print(f"Solo hay {libro['stock']} unidades disponibles.")
        return False

    # Confirmación
    decision = input(f"¿Añadir '{libro['titulo']}' x {cantidad} al carrito? (Si/No): ").strip().lower()
    if decision not in ("si", "sí", "si ", "s", "yes", "y", "no", "n"):
        print("No logré entender tu respuesta. Por favor escribe 'Si' o 'No'.")
        return False
    if decision in ("no","n"):
        print("No se añadió nada al carrito.")
        return False

    # Cálculo 
    precio_unit = float(libro["precio"])
    subtotal = round(precio_unit * cantidad, 2)

    # Actualiza stock e inserta al carrito
    libro["stock"] -= cantidad
    carrito.append({
        "titulo": libro["titulo"],
        "autor":  libro["autor"],
        "precio_unit": precio_unit,
        "cantidad": cantidad,
        "subtotal": subtotal
    })

    print(f"Añadido: '{libro['titulo']}' x{cantidad} | "
          f"Total ítem: ${subtotal:,.2f}".replace(",", "."))
    return True    
```

## 5. Uso de bucle while para iterar hasta que el usuario decida salir (1 punto):
- Implementa un bucle while que permita al usuario realizar múltiples compras hasta
que ingrese una opción de salida.


```python
# Función ver_carrito
def ver_carrito():
    if not carrito:
        print("Tu carrito está vacío.")
        return

    total = sum(item.get("subtotal", 0.0) for item in carrito)
    total_ahorro = sum(item.get("ahorro_item", 0.0) for item in carrito)

    print("Carrito actual:")
    for it in carrito:
        desc_txt = f" | Desc: {it['descuento_autor']*100:.2f}%" if it.get("descuento_autor", 0) else ""
        print(f"- {it['titulo']} x{it['cantidad']} "
              f" ${it['precio_unit']:.2f} = ${it['subtotal']:.2f}{desc_txt}")
    print(f"Total parcial: ${total:.2f}")
    if total_ahorro > 0:
        print(f"Ahorro acumulado por descuentos: ${total_ahorro:.2f}")
```

```python
# Función pagar
def pagar():
    if not carrito:
        print("No puedes pagar con el carrito vacío.")
        return
    print("Procesando pago... (simulado)")
    vaciar_carrito()  # limpia carrito tras pagar
```

## 7. Simulación de una factura (2 punto):
- Al finalizar la compra, muestra un resumen con el total de libros comprados, el monto
total pagado y el ahorro por descuentos.



```python
# Factura detallada

def generar_factura():
    if not carrito:
        print("No hay elementos en el carrito. No se puede generar la factura.")
        return

    total_pagado = sum(it.get("subtotal", 0.0) for it in carrito)
    total_ahorro = sum(it.get("ahorro_item", 0.0) for it in carrito)
    total_libros = sum(it.get("cantidad", 0) for it in carrito)
    print("Libros & Bytes ")
    print("=" * 40)
    print("FACTURA")
    print("=" * 40)
    for it in carrito:
        desc_txt = f" (desc {it['descuento_autor']*100:.2f}%)" if it.get("descuento_autor", 0) else ""
        print(f"{it['titulo']} x{it['cantidad']}  ${it['precio_unit']:.2f}{desc_txt} -> ${it['subtotal']:.2f}")
    print("-" * 40)
    print(f"Libros totales        : {total_libros}")
    print(f"Monto pagado          : ${total_pagado:.2f}")
    print(f"Ahorro por descuentos : ${total_ahorro:.2f}")
    print("=" * 40)
    print("¡Gracias por su compra!")

# Flujo sugerido al pagar
def pagar():
    if not carrito:
        print("No puedes pagar con el carrito vacío.")
        return
    print("Procesando pago... (simulado)")
    generar_factura()
    vaciar_carrito()  # limpia carrito tras pagar
```

```python
# Descuentos por autor

# Diccionario con los descuentos por autor (puedes editarlo según lo que pida tu pauta)
descuentos_por_autor = {
    "Gabriel García Márquez": 0.10,  # 10%
    "Jorge Luis Borges": 0.05,       # 5%
    "juan":0.5 #50%
}

def comprar_libros(titulo: str, cantidad: int):

    if not isinstance(cantidad, int):
        print("La cantidad debe ser un entero.")
        return False

    libro = next((l for l in inventario if l["titulo"].lower() == titulo.strip().lower()), None)
    if not libro:
        print("No encontramos ese título en el inventario.")
        return False
    if cantidad == 0:
        print("¿Me está diciendo que quiere comprar 0 libros?, no hay chance!")
        return False
    if cantidad < 0:
        print(f"jajaja, Difícil comprar {cantidad} libros.")
        return False

    if libro["stock"] <= 0:
        print("No hay stock disponible de este título.")
        return False
    if cantidad > libro["stock"]:
        print(f"Solo hay {libro['stock']} unidades disponibles.")
        return False

    # Confirmación
    decision = input(f"¿Añadir '{libro['titulo']}' x{cantidad} al carrito? (Si/No): ").strip().lower()
    if decision in ("no", "n"):
        print("No se añadió nada al carrito.")
        return False
    if decision not in ("si", "sí", "s", "y", "yes"):
        print("No logré entender tu respuesta. Por favor escribe 'Si' o 'No'.")
        return False

    # Aplicar descuento por autor si corresponde
    precio_base = float(libro["precio"])
    descuento = float(descuentos_por_autor.get(libro["autor"], 0.0))
    precio_unit = round(precio_base * (1 - descuento), 2)

    subtotal = round(precio_unit * cantidad, 2)
    ahorro_item = round((precio_base - precio_unit) * cantidad, 2)

    # Actualiza stock e inserta al carrito
    libro["stock"] -= cantidad
    carrito.append({
        "titulo": libro["titulo"],
        "autor":  libro["autor"],
        "precio_unit": precio_unit,
        "descuento_autor": descuento,
        "ahorro_item": ahorro_item,
        "cantidad": cantidad,
        "subtotal": subtotal
    })

    if descuento > 0:
        print(f"¡Descuento aplicado por autor ({descuento*100:.2f}%)!")
    print(f"Añadido: '{libro['titulo']}' x{cantidad} | "
          f"Total ítem: ${subtotal:,.2f} | Ahorro: ${ahorro_item:,.2f}".replace(",", "."))
    return True
```

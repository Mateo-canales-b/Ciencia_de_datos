```python
# Módulo 5 sesión 5
 # a) IC para la media (σ conocida)
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt
import math

# Datos
x_barra = 50     # media muestral
sigma = 10       # desviación estándar
n = 36           # tamaño de muestra
z = norm.ppf(0.975)  # valor z para 95% de confianza

# Cálculo del intervalo de confianza
error = z * sigma / np.sqrt(n)
IC = (x_barra - error, x_barra + error)
print(f"95% IC para la media (σ conocida): {IC}")
```

```python
# Intervalo de confianza para la media con sigma desconocida
from scipy.stats import t

x_barra = 50
s = 10
n = 20
t_crit = t.ppf(0.975, df=n-1)

error = t_crit * s / np.sqrt(n)
IC = (x_barra - error, x_barra + error)
print(f"95% IC para la media (σ desconocida): {IC}")
```

```python
# INtervalo de confianza para proporciones
p_hat = 0.4
n = 100
z = norm.ppf(0.975)

error = z * np.sqrt(p_hat * (1 - p_hat) / n)
IC = (p_hat - error, p_hat + error)
print(f"95% IC para una proporción: {IC}")
```

```python
# Prueba de hipótesis para la media (Test Z)
# H0: mu=50000
# H1: mu distinto de 50000
mu = 50000
x_bar = 48000
sigma = 1000
n = 500
z = (x_bar - mu) / (sigma / np.sqrt(n)) # sigue una distribución normal estándar bajo el supuesto de que H0 es verdadera
p_value = 2 * (1 - norm.cdf(abs(z)))

print(f"Z = {z:.2f}, valor-p = {p_value:.4f}")
```

```python
# Tamaño de Muestra para Estimar la Media
import math

sigma = 10
E = 2
z = norm.ppf(0.975)

n = math.ceil((z * sigma / E)**2)
print(f"Tamaño de muestra necesario: {n}")
```

```python
# Simulación para Validar el Teorema del Límite Central (TLC)
import matplotlib.pyplot as plt

mu, sigma = 50, 10
n = 36
muestras = 1000

medias = [np.mean(np.random.normal(mu, sigma, n)) for _ in range(muestras)]
plt.hist(medias, bins=30, density=True, alpha=0.6, color='blue')
plt.title("Distribución muestral de la media (TLC)")
plt.xlabel("Media muestral")
plt.ylabel("Densidad")
plt.grid(True)
plt.show()
```

```python
# Otros ejemplos con sigma conocida
import numpy as np
from scipy.stats import norm

# Datos
x_bar = 50       # Media muestral
sigma = 10       # Desviación estándar poblacional
n = 36           # Tamaño de la muestra
confianza = 0.95 # Nivel de confianza

# Calcular el valor Z crítico
z_critico = norm.ppf(1 - (1 - confianza) / 2)

# Calcular el margen de error
margen_error = z_critico * (sigma / np.sqrt(n))

# Calcular los límites del intervalo
limite_inferior = x_bar - margen_error
limite_superior = x_bar + margen_error

print(f"Valor crítico Z: {z_critico:.2f}")
print(f"Margen de error: {margen_error:.2f}")
print(f"Intervalo de confianza ({confianza*100:.0f}%): [{limite_inferior:.2f}, {limite_superior:.2f}]")
```

```python
# Intervalo de confianza para la proporción
from scipy.stats import norm
import numpy as np

# Datos
n = 100           # Tamaño de la muestra
p_hat = 40/100    # Proporción muestral (40 éxitos de 100)

# Calcular el error estándar
error_estandar = np.sqrt(p_hat * (1 - p_hat) / n)

# Calcular el valor Z crítico para 95% de confianza (dos colas)
z_critico = norm.ppf(0.975)

# Calcular los límites del intervalo
limite_inferior = p_hat - z_critico * error_estandar
limite_superior = p_hat + z_critico * error_estandar

print(f"Proporción muestral: {p_hat:.2f}")
print(f"Intervalo de confianza (95%): [{limite_inferior:.2f}, {limite_superior:.2f}]")
```

```python
# Otro ejemplo calculo del tamaño de la muestra
import math
from scipy.stats import norm

# Datos
sigma = 10      # Desviación estándar poblacional
E = 2           # Margen de error deseado
confianza = 0.95 # Nivel de confianza

# Calcular el valor Z crítico
z_critico = norm.ppf(1 - (1 - confianza) / 2)

# Calcular el tamaño muestral
n_calculado = (z_critico * sigma / E)**2

# Redondear hacia arriba
n = math.ceil(n_calculado)

print(f"Valor crítico Z: {z_critico:.2f}")
print(f"Tamaño muestral calculado: {n_calculado:.2f}")
print(f"Tamaño muestral requerido (redondeado): {n}")
```

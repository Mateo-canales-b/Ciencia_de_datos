Ejemplo 1: Aplicación del Método Científico en Ciencia de Datos

Un analista de datos quiere saber si un nuevo diseño de página web aumenta el tiempo que los usuarios pasan en ella.


Problema: ¿El nuevo diseño de la página web aumenta el tiempo de permanencia de los usuarios?


Hipótesis: El nuevo diseño de la página web incrementará el tiempo promedio que los usuarios pasan en el sitio.

Experimentación: Se divide a los visitantes del sitio web en dos grupos de forma aleatoria: el Grupo A ve el diseño actual y el Grupo B ve el nuevo diseño. Se registra el tiempo de permanencia de cada usuario en ambos grupos durante un mes.


Análisis: Se utilizan pruebas estadísticas (por ejemplo, una prueba t de Student) para comparar el tiempo promedio de permanencia entre el Grupo A y el Grupo B.

Conclusión: Si la prueba estadística muestra una diferencia significativa y el Grupo B tuvo un tiempo de permanencia mayor, se acepta la hipótesis de que el nuevo diseño mejora el tiempo de permanencia. En caso contrario, se rechaza o se replantea la hipótesis.

Ejemplo 2: Importancia del Pensamiento Estadístico

Una empresa de comercio electrónico lanza una nueva campaña de marketing y quiere saber si ha sido efectiva para aumentar las ventas.

Sin pensamiento estadístico: La empresa observa que las ventas aumentaron un 5% después de la campaña y concluye que la campaña fue un éxito. Sin embargo, no considera otros factores como una temporada de descuentos o el lanzamiento de un nuevo producto.

Con pensamiento estadístico: La empresa recolecta datos de ventas antes y después de la campaña, y también de un grupo de control (clientes que no fueron expuestos a la campaña). Utiliza pruebas de hipótesis para determinar si el aumento en las ventas del grupo expuesto a la campaña es estadísticamente significativo y no se debe al azar o a otras variables. Esto le permite tomar una decisión informada sobre la efectividad de la campaña y si debe invertir más en ella.

Ejemplo 3: Inferencia Estadística en la Vida Cotidiana

Quieres saber la opinión promedio de los estudiantes de tu universidad sobre las nuevas instalaciones deportivas, pero no puedes preguntar a todos.


Población: Todos los estudiantes de la universidad.


Muestra: Seleccionas aleatoriamente a 200 estudiantes de diferentes facultades y les aplicas una encuesta.

Inferencia: Basándote en las respuestas de esos 200 estudiantes (la muestra), calculas el promedio de satisfacción con las instalaciones. Luego, utilizas la inferencia estadística para generalizar este resultado y estimar la satisfacción promedio de

todos los estudiantes de la universidad, con un cierto nivel de confianza.


Aplicación: Si la mayoría de los estudiantes encuestados expresan insatisfacción, la universidad puede inferir que las nuevas instalaciones necesitan mejoras para la población estudiantil en general.

Ejemplo 4: El Concepto de Significancia (Valor P)

Un equipo de investigación está probando un nuevo fertilizante para plantas y quiere ver si realmente aumenta el crecimiento.


Hipótesis Nula (H0): El nuevo fertilizante no tiene ningún efecto en el crecimiento de las plantas (es decir, el crecimiento es el mismo que sin fertilizante).

Experimento: Aplican el nuevo fertilizante a un grupo de plantas y a otro grupo similar no le aplican nada (grupo de control). Después de un mes, miden el crecimiento promedio de ambos grupos.

Resultados y Valor P: Supongamos que encuentran que las plantas con el nuevo fertilizante crecieron un poco más. Calculan el p-valor.

Si el p-valor es 0.01 (p<0.05): Esto significa que hay solo un 1% de probabilidad de que esa diferencia en el crecimiento se haya producido por puro azar si el fertilizante no tuviera efecto. Dado que es una probabilidad muy baja, los investigadores rechazan la hipótesis nula y concluyen que el fertilizante sí tiene un efecto estadísticamente significativo en el crecimiento.

Si el p-valor es 0.20 (p>0.05): Esto significa que hay un 20% de probabilidad de que la diferencia observada se haya producido por azar. Como esta probabilidad no es tan baja, los investigadores no pueden rechazar la hipótesis nula. Concluyen que no hay suficiente evidencia para afirmar que el fertilizante tiene un efecto significativo.

Ejemplo 5: Diseño de Experimentos A/B en Marketing

Una aplicación de delivery quiere saber qué versión de su botón de "Pedir Ahora" (uno verde o uno rojo) genera más clics.


Objetivo: Maximizar los clics en el botón "Pedir Ahora".


Hipótesis: El botón rojo generará más clics que el botón verde.

Variables:


Independiente: El color del botón (verde o rojo).


Dependiente: El número de clics en el botón.

Grupos:


Grupo Experimental: Un porcentaje de usuarios ve el botón rojo.


Grupo de Control: Otro porcentaje de usuarios ve el botón verde.


Aleatorización: Los usuarios son asignados aleatoriamente a ver el botón verde o rojo para evitar sesgos.


Replicación: El experimento se ejecuta durante varias semanas para asegurar que los resultados sean consistentes y no se deban a variaciones temporales.

Análisis: Se comparan las tasas de clics de ambos grupos y se utiliza una prueba estadística para determinar si la diferencia es significativa. Si el botón rojo tiene una tasa de clics significativamente más alta, la empresa decide implementarlo.

Planteamiento del Problema e Hipótesis

Concepto: Reitera la importancia de una pregunta clara y una hipótesis comprobable (nula y alternativa).


Enseñanza con Python:

Simulación de datos: Como el problema de las horas de estudio y calificaciones. Puedes simular estos datos iniciales en Python para tener algo con qué trabajar.



```python
import pandas as pd # Para manejar datos en tablas
import numpy as np  # Para operaciones numéricas
from scipy.stats import pearsonr # Para calcular la correlación de Pearson
from scipy.stats import ttest_ind # Para la prueba t en el ejemplo de A/B Testing

# Paso 1: Planteamiento del Problema
problema = "¿El tiempo de estudio afecta las calificaciones de los estudiantes?"
print(f"Problema a investigar: {problema}\n")

# Paso 2: Formulación de Hipótesis
# Basado en el ejemplo: A mayor tiempo de estudio, mejores calificaciones
hipotesis_nula = "H0: El tiempo de estudio NO afecta las calificaciones de los estudiantes."
hipotesis_alternativa = "H1: A mayor tiempo de estudio, mejores calificaciones."

print(f"Hipótesis Nula: {hipotesis_nula}")
print(f"Hipótesis Alternativa: {hipotesis_alternativa}\n")

# Simulamos algunos datos para el ejemplo
np.random.seed(42) # Para reproducibilidad
horas_estudio_simuladas = np.random.randint(10, 50, 100) # Entre 10 y 50 horas para 100 estudiantes
# Calificaciones_simuladas tienen una relación positiva con horas_estudio_simuladas, más algo de ruido
calificaciones_simuladas = horas_estudio_simuladas * 1.5 + np.random.normal(0, 10, 100)
# Asegurarse de que las calificaciones no superen 100
calificaciones_simuladas = np.clip(calificaciones_simuladas, 0, 100).astype(int)

df_simulado = pd.DataFrame({
    'Horas_Estudio': horas_estudio_simuladas,
    'Calificaciones': calificaciones_simuladas
})

print("Primeros 5 registros de nuestros datos simulados:")
print(df_simulado.head())
```

```python
# Paso 3: Definir las Variables
variable_independiente = "Horas de estudio por semana"
variable_dependiente = "Calificación final en el curso"

print(f"\nVariable Independiente: {variable_independiente}")
print(f"Variable Dependiente: {variable_dependiente}\n")
```

```python
# Paso 4: Diseño del Experimento y Recolección de Datos (en este caso, simulada)
print("Hemos simulado 100 registros de datos de 'Horas_Estudio' y 'Calificaciones'.")
print("En un experimento real, estos datos se recolectarían de una muestra representativa.")

print("\nConsideraciones de control de variables (explicar teóricamente):")
print("- Asegurar acceso a mismos recursos educativos para todos.")
print("- Excluir casos atípicos (enfermedades prolongadas, etc.).")
```

```python
# Paso 5: Realizar el Análisis de Datos
print("\nAnálisis Descriptivo de los datos:")
print(df_simulado.describe()) # Estadísticas básicas como media, desviación estándar
```

```python
# Calcular la Correlación de Pearson
correlacion, p_valor = pearsonr(df_simulado['Horas_Estudio'], df_simulado['Calificaciones'])

print(f"\nResultado del Análisis Estadístico (Correlación de Pearson):")
print(f"Correlación: {correlacion:.2f}")
print(f"P-valor: {p_valor:.3f}")
```

```python
# Paso 6: Interpretar los Resultados
print("\nInterpretación de los Resultados:")
print(f"La correlación de {correlacion:.2f} indica la fuerza y dirección de la relación lineal.")
if correlacion > 0:
    print("Es una correlación positiva, lo que sugiere que a medida que las horas de estudio aumentan, las calificaciones tienden a aumentar.")
elif correlacion < 0:
    print("Es una correlación negativa, lo que sugiere que a medida que las horas de estudio aumentan, las calificaciones tienden a disminuir.")
else:
    print("La correlación es cercana a cero, lo que sugiere poca o ninguna relación lineal.")

print(f"El p-valor de {p_valor:.3f} nos ayuda a determinar la significancia estadística.")

if p_valor < 0.05:
    print("Como el p-valor es menor que 0.05, el resultado es estadísticamente significativo.")
    print("Esto significa que es poco probable que la relación observada se deba al azar.")
    decision = "Rechazamos la Hipótesis Nula (H0)."
    concluir_hipotesis = "Aceptamos la Hipótesis Alternativa (H1): A mayor tiempo de estudio, mejores calificaciones."
else:
    print("Como el p-valor es mayor o igual que 0.05, el resultado NO es estadísticamente significativo.")
    print("Esto significa que la relación observada podría deberse al azar.")
    decision = "No podemos rechazar la Hipótesis Nula (H0)."
    concluir_hipotesis = "No hay suficiente evidencia para afirmar que el tiempo de estudio afecta significativamente las calificaciones."

print(f"\nDecisión Estadística: {decision}")
print(f"Conclusión de la Hipótesis: {concluir_hipotesis}\n")
```

```python
# Paso 7: Concluir y Comunicar los Hallazgos
print("Conclusión final basada en nuestro experimento:")
print(f"Basándonos en los datos simulados y el análisis estadístico, {concluir_hipotesis.lower()}.")
print("Este ejercicio demuestra cómo el método científico, apoyado por Python y la estadística, nos permite tomar decisiones basadas en evidencia en ciencia de datos.")

print("\nMétodos de Investigación: Cualitativos vs. Cuantitativos")
print("Python es una herramienta fundamental para los métodos cuantitativos, ya que nos permite trabajar con datos numéricos y aplicar estadísticas.")

print("\nPara métodos cualitativos (ej. entrevistas), Python podría usarse para:")
print("- Análisis de sentimientos de texto.")
print("- Identificación de temas clave en grandes volúmenes de texto (con PNL).")
print("Sin embargo, la recolección inicial de estos datos es diferente a la numérica.")

print("\nDiseño de Experimentos: Ejemplo de A/B Testing en Ciencia de Datos")
```

```python
np.random.seed(42)

ctr_grupo_a = np.random.normal(loc=0.03, scale=0.005, size=500)
ctr_grupo_a = np.clip(ctr_grupo_a, 0.01, 0.05)

ctr_grupo_b = np.random.normal(loc=0.035, scale=0.005, size=500)
ctr_grupo_b = np.clip(ctr_grupo_b, 0.01, 0.06)

print(f"CTR Promedio Algoritmo A (Control): {np.mean(ctr_grupo_a):.4f}")
print(f"CTR Promedio Algoritmo B (Experimental): {np.mean(ctr_grupo_b):.4f}")

t_stat, p_val_ab = ttest_ind(ctr_grupo_a, ctr_grupo_b)

print(f"\nResultado de la prueba t entre Algoritmo A y B:")
print(f"Estadístico t: {t_stat:.2f}")
print(f"P-valor: {p_val_ab:.4f}")

if p_val_ab < 0.05:
  # Rechazando H0: media 1=media2
    print("El p-valor es menor que 0.05, por lo que la diferencia en CTR entre los algoritmos es estadísticamente significativa.")
    print("Esto sugiere que el Algoritmo B realmente tiene un impacto diferente en el CTR.")
else:
    print("El p-valor es mayor que 0.05, por lo que la diferencia en CTR no es estadísticamente significativa.")
    print("No hay suficiente evidencia para afirmar que el Algoritmo B es mejor que el A basándonos en estos datos.")
```

```python
print("\nComponentes del Diseño de Experimentos aplicados:")
print(f"- Hipótesis: El nuevo algoritmo de clasificación mejora la precisión de un modelo.")
print(f"- Variables: Algoritmo (Independiente), Precisión del modelo (Dependiente).")
print(f"- Grupos: Algoritmo actual (Control), Nuevo algoritmo (Experimental).")
print(f"- Aleatorización: Asignación aleatoria de usuarios a los grupos (implícito en la simulación).")
print(f"- Replicación: El experimento debería repetirse o ejecutarse con suficientes datos para asegurar consistencia.")

print("\nRecapitulación: El Método Científico y la Estadística con Python")
print("Hemos visto cómo Python nos permite aplicar cada paso del método científico:")
print("- Definir un problema y formular una hipótesis.")
print("- Simular y preparar datos para el análisis.")
print("- Realizar análisis estadísticos (descriptivos e inferenciales).")
print("- Interpretar resultados y sacar conclusiones basadas en evidencia (p-valor).")
print("- Diseñar experimentos para probar ideas de forma rigurosa.")

print("\n¡En ciencia de datos, estas herramientas son esenciales para pasar de la intuición a las decisiones basadas en datos!")
```

```python
# Ejemplo de Tipos de Muestreo en Python
# ================================

import pandas as pd
import numpy as np
import random

# Paso 1: Crear un DataFrame simulado
np.random.seed(42)
n = 100
df = pd.DataFrame({
    'ID': range(1, n+1),
    'edad': np.random.randint(18, 70, size=n),
    'genero': np.random.choice(['Masculino', 'Femenino'], size=n),
    'region': np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], size=n)
})

print("Primeros datos del marco muestral:")
print(df.head())
```

```python
# 1. Muestreo Aleatorio Simple (MAS)
# ======================================
muestra_mas = df.sample(n=10, random_state=1)
print("\nMuestreo Aleatorio Simple:")
print(muestra_mas)
```

```python
# 2. Muestreo Sistemático
# ======================================
# Tomamos cada k-ésimo elemento
k = 10
indices = np.arange(0, len(df), k)
muestra_sistematico = df.iloc[indices]
print("\nMuestreo Sistemático:")
print(muestra_sistematico)
```

```python
# 3. Muestreo Estratificado
# ======================================
# Estratificamos por género
from sklearn.model_selection import train_test_split

muestra_estratificada = df.groupby('genero', group_keys=False).apply(lambda x: x.sample(frac=0.1, random_state=42))
# df.groupby('genero'):
# Agrupa el DataFrame por el valor de la columna genero.
# Ejemplo: dos grupos → uno para "Masculino" y otro para "Femenino".
print("\nMuestreo Estratificado por género:")
print(muestra_estratificada)
# apply(lambda x: x.sample(frac=0.1, random_state=42)):

# Para cada grupo (x), toma una muestra aleatoria del 10% (frac=0.1) de ese grupo.

# Usa random_state=42 para asegurar que el muestreo sea reproducible.
# group_keys=False:

# Hace que el índice de los grupos no se agregue al índice final del DataFrame.

# Esto mantiene el formato más limpio.

```

```python
# 4. Muestreo por Conglomerados
# ======================================
# Suponemos que la región es el conglomerado, seleccionamos 1 al azar
regiones = df['region'].unique()
region_elegida = np.random.choice(regiones, 1)[0]
# Usa np.random.choice() para seleccionar una región al azar del arreglo anterior.
# [0] es para acceder directamente al valor (ya que choice devuelve una lista si se pone size=1).
# Ejemplo: region_elegida = 'Centro'
muestra_conglomerado = df[df['region'] == region_elegida]

print(f"\nMuestreo por Conglomerados - Región seleccionada: {region_elegida}")
print(muestra_conglomerado.head())
# Se extraen todas las categorías únicas de la columna region del DataFrame df.
# Ejemplo de salida: ['Norte', 'Centro', 'Sur']
```

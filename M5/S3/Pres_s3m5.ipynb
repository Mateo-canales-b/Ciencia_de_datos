```python
# Módulo 5 sesión 3
# 1. Variable Aleatoria Discreta - Distribución Binomial
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros
n = 10    # número de ensayos
p = 0.5   # probabilidad de éxito

x = np.arange(0, n+1)
prob = binom.pmf(x, n, p)

# Gráfico (sin use_line_collection)
plt.stem(x, prob)
plt.title('Distribución Binomial: n=10, p=0.5')
plt.xlabel('Número de éxitos')
plt.ylabel('Probabilidad')
plt.grid(True)
plt.show()
```

```python
# 2. Variable Aleatoria Discreta - Distribución de Poisson
from scipy.stats import poisson

# Parámetro
lam = 4  # tasa de eventos por intervalo

x = np.arange(0, 15)
prob = poisson.pmf(x, mu=lam)

plt.bar(x, prob)
plt.title('Distribución de Poisson: λ=4')
plt.xlabel('Número de eventos')
plt.ylabel('Probabilidad')
plt.grid(True)
plt.show()
```

```python
# 3. Variable Aleatoria Continua - Distribución Normal
from scipy.stats import norm

# Parámetros
mu = 65
sigma = 5
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 100)

plt.plot(x, norm.pdf(x, mu, sigma), label='Normal(65, 5)')
plt.fill_between(x, norm.pdf(x, mu, sigma), alpha=0.3)
plt.title('Distribución Normal')
plt.xlabel('x')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 4. Distribución Exponencial - Tiempo entre eventos
from scipy.stats import expon

# Parámetro: tasa de ocurrencia λ
lambd = 1/5  # un evento cada 5 minutos
x = np.linspace(0, 30, 100)
pdf = expon.pdf(x, scale=1/lambd)

plt.plot(x, pdf, label='Exponencial (λ=1/5)')
plt.fill_between(x, pdf, alpha=0.3)
plt.title('Distribución Exponencial')
plt.xlabel('Tiempo')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# 5. Distribución Normal Estándar - Cálculo de Probabilidades
# P(X <= 70) si X ~ N(65, 5)
z = (70 - 65) / 5
probabilidad = norm.cdf(z)

print(f"P(X <= 70) = {probabilidad:.4f}")  # Resultado: ~0.8413
```

```python
# 6. Probabilidad entre dos valores con Normal
# P(60 <= X <= 70) para N(65, 5)
z1 = (60 - 65) / 5
z2 = (70 - 65) / 5
prob = norm.cdf(z2) - norm.cdf(z1)

print(f"P(60 <= X <= 70) = {prob:.4f}")  # Resultado: ~0.6826
```

```python
# 7. Distribución Uniforme Continua
from scipy.stats import uniform

a, b = 0, 60  # intervalo de tiempo
x = np.linspace(a, b, 100)
pdf = uniform.pdf(x, loc=a, scale=b-a)

plt.plot(x, pdf)
plt.fill_between(x, pdf, alpha=0.3)
plt.title('Distribución Uniforme Continua entre 0 y 60')
plt.xlabel('x')
plt.ylabel('Densidad')
plt.grid(True)
plt.show()
```

```python
# Ejemplo: El número de caras obtenidas al lanzar una moneda tres veces (valores posibles: 0, 1, 2, 3).

import random

def lanzar_moneda_discreta(num_lanzamientos):
    """Simula el número de caras en un número dado de lanzamientos de moneda."""
    caras = 0
    for _ in range(num_lanzamientos):
        if random.random() > 0.5: # 0.5 de probabilidad para cara (éxito)
            caras += 1
    return caras

# Ejemplo: Lanzar una moneda 3 veces
resultados = [lanzar_moneda_discreta(3) for _ in range(10)] # Simular 10 veces el experimento
print("Resultados del número de caras en 3 lanzamientos:", resultados)
```

```python
# Ejemplo: El tiempo de espera en una fila (puede ser 2 minutos, 2.5 minutos, 3.78 minutos, etc.)
import random

def tiempo_espera_continua(min_tiempo, max_tiempo):
    """Simula un tiempo de espera continuo dentro de un rango."""
    return random.uniform(min_tiempo, max_tiempo)

# Ejemplo: Tiempo de espera entre 0 y 10 minutos
tiempo = tiempo_espera_continua(0, 10)
print("Tiempo de espera simulado (minutos):", round(tiempo, 2))
```

```python
# Ejemplo: Lanzamiento de un dado justo de seis caras, donde cada número tiene una probabilidad de 1/6.
def funcion_probabilidad_dado(x):
    """Función de probabilidad para un dado justo."""
    if 1 <= x <= 6:
        return 1/6
    else:
        return 0

# Ejemplo: Probabilidad de obtener un 3
prob_3 = funcion_probabilidad_dado(3)
print(f"Probabilidad de obtener un 3 en un dado: {prob_3}")

# Verificar que la suma de probabilidades es 1
suma_probs = sum(funcion_probabilidad_dado(i) for i in range(1, 7))
print(f"Suma de todas las probabilidades: {suma_probs}")
```

```python
# Ejemplo: Estatura de adultos que se distribuye normalmente con media y desviación estándar
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt

# Ejemplo de FDP Normal
mu = 170  # cm [cite: 301]
sigma = 10 # cm [cite: 301]

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
pdf_values = norm.pdf(x, mu, sigma)

plt.figure(figsize=(8, 4))
plt.plot(x, pdf_values, label=f'FDP Normal (μ={mu}, σ={sigma})')
plt.title('Función de Densidad de Probabilidad Normal')
plt.xlabel('Estatura (cm)')
plt.ylabel('Densidad de Probabilidad')
plt.grid(True)
plt.legend()
plt.show()

# Probabilidad de que la estatura esté entre 160 y 180 cm
prob_interval = norm.cdf(180, mu, sigma) - norm.cdf(160, mu, sigma)
print(f"Probabilidad de que la estatura esté entre 160 y 180 cm: {prob_interval:.4f}")
```

```python
# Función de Distribución Acumulada (CDF) describe la probabilidad acumulada de que una variable aleatoria X tome
# un valor menor o igual a un valor específico x.
# Ejemplo con Variable Discreta: Número de caras al lanzar una moneda dos veces.
def cdf_moneda_discreta(x):
    """CDF para el número de caras al lanzar dos monedas."""
    if x < 0:
        return 0
    elif 0 <= x < 1:
        return 1/4 # P(X=0)
    elif 1 <= x < 2:
        return 1/4 + 2/4 # P(X=0) + P(X=1)
    else: # x >= 2
        return 1 # P(X=0) + P(X=1) + P(X=2)

print(f"F(0) = P(X <= 0): {cdf_moneda_discreta(0)}")
print(f"F(1) = P(X <= 1): {cdf_moneda_discreta(1)}")
print(f"F(2) = P(X <= 2): {cdf_moneda_discreta(2)}")
```

```python
# Ejemplo con Variable Continua (Distribución Exponencial): Tiempo de espera en una fila que sigue una distribución exponencial con parámetro λ=1
from scipy.stats import expon

# Parámetro lambda
lambda_param = 1

# CDF para la distribución exponencial
def cdf_exponencial(x, lambda_val):
    return 1 - np.exp(-lambda_val * x)

# Probabilidad de esperar menos de 3 minutos
prob_less_than_3 = cdf_exponencial(3, lambda_param) # Using the formula
print(f"Probabilidad de que el tiempo de espera sea menor a 3 minutos: {prob_less_than_3:.4f}")

# Usando scipy.stats
prob_less_than_3_scipy = expon.cdf(3, scale=1/lambda_param) # scale is 1/lambda for scipy's expon
print(f"Probabilidad (usando scipy): {prob_less_than_3_scipy:.4f}")
```

```python
# Ejemplo: Lanzamiento de un dado justo, donde cada número tiene una probabilidad de 1/6.
from scipy.stats import randint

# Un dado justo de 6 caras
min_val = 1
max_val = 6

# PMF para un valor específico (e.g., 3)
prob_of_3 = randint.pmf(3, min_val, max_val + 1) # max_val is exclusive in randint
print(f"Probabilidad de obtener un 3 (Uniforme Discreta): {prob_of_3}")

# Todos los posibles resultados
valores = list(range(min_val, max_val + 1))
probabilidades = [randint.pmf(val, min_val, max_val + 1) for val in valores]
print(f"Probabilidades para cada cara del dado: {dict(zip(valores, probabilidades))}")
```

```python
# Ejemplo: Lanzar una moneda sesgada donde la probabilidad de cara (éxito) es 0.7 y de cruz (fracaso) es 0.3.
from scipy.stats import bernoulli

p_exito = 0.7 # Probabilidad de cara

# Probabilidad de obtener un éxito (cara)
prob_cara = bernoulli.pmf(1, p_exito)
print(f"Probabilidad de obtener cara (Bernoulli, p=0.7): {prob_cara}")

# Probabilidad de obtener un fracaso (cruz)
prob_cruz = bernoulli.pmf(0, p_exito)
print(f"Probabilidad de obtener cruz (Bernoulli, p=0.7): {prob_cruz}")
```

```python
# Ejemplo: Lanzar 10 veces una moneda equilibrada (p=0.5), la probabilidad de obtener exactamente 4 caras.
# from scipy.stats import binom

n_ensayos = 10  # Número total de lanzamientos
k_exitos = 4    # Número de caras deseadas
p_prob_exito = 0.5 # Probabilidad de cara en cada lanzamiento (moneda equilibrada)

# Probabilidad de obtener exactamente k éxitos
prob_4_caras = binom.pmf(k_exitos, n_ensayos, p_prob_exito)
print(f"Probabilidad de obtener exactamente 4 caras en 10 lanzamientos: {prob_4_caras:.4f}")
```

```python
# Ejemplo: Un call center recibe en promedio 5 llamadas por hora (λ=5), la probabilidad de recibir exactamente 3 llamadas en una hora.
from scipy.stats import poisson

lambda_tasa = 5 # Promedio de llamadas por hora
k_eventos = 3   # Número de llamadas deseadas

# Probabilidad de k eventos
prob_3_llamadas = poisson.pmf(k_eventos, lambda_tasa)
print(f"Probabilidad de recibir exactamente 3 llamadas en una hora: {prob_3_llamadas:.4f}")
```

```python
# Ejemplo: El tiempo de espera en un semáforo está distribuido uniformemente entre 0 y 60 segundos.
from scipy.stats import uniform

a = 0  # Límite inferior
b = 60 # Límite superior

# La FDP es constante 1/(b-a) en el intervalo [a, b]
fdp_valor = uniform.pdf(30, loc=a, scale=b-a) # FDP en el punto 30 (cualquier punto dentro del rango)
print(f"Densidad de probabilidad para un tiempo de espera de 30 segundos: {fdp_valor:.4f}")

# Probabilidad de esperar entre 10 y 20 segundos
prob_intervalo = uniform.cdf(20, loc=a, scale=b-a) - uniform.cdf(10, loc=a, scale=b-a)
print(f"Probabilidad de esperar entre 10 y 20 segundos: {prob_intervalo:.4f}")
```

```python
# Ejemplo: Calcular P(X ≤ 70) para una normal con μ = 65 y σ = 5
from scipy.stats import norm

mu = 65 # Media
sigma = 5 # Desviación estándar
x_valor = 70 # Valor a evaluar

# P(X <= 70)
prob_le_70 = norm.cdf(x_valor, loc=mu, scale=sigma)
print(f"P(X <= 70) para N(μ=65, σ=5): {prob_le_70:.4f}")

# P(X >= 70)
prob_ge_70 = 1 - prob_le_70
print(f"P(X >= 70) para N(μ=65, σ=5): {prob_ge_70:.4f}")
```

```python
# Ejemplo: Calcular P(Z ≤ 1.96).
from scipy.stats import norm

z_valor = 1.96

# P(Z <= 1.96)
prob_le_1_96 = norm.cdf(z_valor, loc=0, scale=1) # media=0, std=1 para normal estándar
print(f"P(Z <= 1.96) para Normal Estándar: {prob_le_1_96:.4f}")

# P(Z >= 1.96)
prob_ge_1_96 = 1 - prob_le_1_96
print(f"P(Z >= 1.96) para Normal Estándar: {prob_ge_1_96:.4f}")
```

```python
# Ejemplo: Clientes llegan a un banco en promedio cada 5 minutos, el tiempo entre llegadas sigue una
# distribución exponencial con λ = 1/5 = 0.2 (ya que lambda es una tasa, si es cada 5 minutos, la tasa es
# 1/5 llegadas por minuto)
from scipy.stats import expon

# Si en promedio llegan clientes cada 5 minutos, la tasa (lambda) es 1/5 = 0.2 eventos por minuto.
# En scipy, `scale` es 1/lambda.
lambda_rate = 1/5 # Tasa de ocurrencia (clientes por minuto)
scale_param = 1 / lambda_rate # Scale parameter for scipy expon

# Probabilidad de que el tiempo entre llegadas sea menor o igual a 3 minutos
time_val = 3
prob_le_3_min = expon.cdf(time_val, scale=scale_param)
print(f"Probabilidad de que el tiempo entre llegadas sea <= 3 minutos: {prob_le_3_min:.4f}")

# Probabilidad de que el tiempo entre llegadas sea mayor a 10 minutos
prob_ge_10_min = 1 - expon.cdf(10, scale=scale_param)
print(f"Probabilidad de que el tiempo entre llegadas sea > 10 minutos: {prob_ge_10_min:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import poisson

# 1. Parámetros
tasa_promedio = 5  # clientes por hora
horas = 10

# 2. Simulación con distribución de Poisson
np.random.seed(42)
clientes_por_hora = np.random.poisson(tasa_promedio, horas)

# 3. Resultados
print("Clientes por hora:", clientes_por_hora)
print("Total de clientes en el día:", clientes_por_hora.sum())

# 4. Gráfico de barras
sns.set(style="whitegrid")
plt.figure(figsize=(10,5))
sns.barplot(x=np.arange(1, horas+1), y=clientes_por_hora, color='skyblue')
plt.title("Número de clientes por hora (simulado con Poisson)")
plt.xlabel("Hora")
plt.ylabel("Número de clientes")
plt.show()

# 5. Distribución teórica de Poisson
x = np.arange(0, 16)
pmf = poisson.pmf(x, mu=tasa_promedio)

plt.figure(figsize=(8,4))
plt.stem(x, pmf, basefmt=" ")
plt.title("Distribución teórica de Poisson (media = 5)")
plt.xlabel("Número de clientes")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.show()
```

```python
# Módulo 5 sesión 4
 # Distribución Muestral
# Es la distribución de un estadístico (por ejemplo, la media) obtenida de muchas muestras de la misma población.
# Ejemplo:
# Si una población tiene media μ = 50 y desviación estándar σ = 10, y tomas muestras de n = 25, la media de las
# medias muestrales sigue siendo 50, pero el error estándar es: 10/raiz(25)
import numpy as np

mu = 50 # Media poblacional
sigma = 10 # Desviación estándar poblacional
n = 25 # Tamaño de muestra
error_estandar = sigma / np.sqrt(n)
print(f"Media: {mu}") # E(Media muestral)= Media poblacional
print(f"Error estándar: {error_estandar}")
```

```python
# Ley de los Grandes Números
# Cuando aumentas el tamaño de la muestra, la media muestral se acerca a la media poblacional.
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
resultados = np.random.choice([0, 1], size=10000)  # 0 = cruz, 1 = cara
promedios = np.cumsum(resultados) / (np.arange(1, 10001))

plt.plot(promedios)
plt.axhline(0.5, color='red', linestyle='--', label='Media poblacional')
plt.title("Ley de los Grandes Números")
plt.xlabel("Número de lanzamientos")
plt.ylabel("Proporción de caras")
plt.legend()
plt.show()
# Eje X: Número de lanzamientos de una moneda (de 1 hasta 10,000).
# Eje Y: Proporción acumulada de caras hasta ese número de lanzamientos.
# Curva azul
# Representa la proporción de caras acumulada a medida que se lanzan más veces la moneda.
# Al principio (pocos lanzamientos), la proporción fluctúa mucho: puede subir o bajar rápidamente.
# A medida que aumentan los lanzamientos, estas fluctuaciones disminuyen.
# Finalmente, la proporción se estabiliza cerca de 0.5, que es el valor esperado teórico para una moneda justa (cara = 0.5).
# Línea roja punteada
# Es la media poblacional teórica, en este caso 0.5, que representa la probabilidad de obtener "cara" al lanzar una moneda justa.
# La Ley de los Grandes Números nos dice que, al aumentar el tamaño de la muestra (número de lanzamientos), la media muestral
 # (proporción de caras observadas) converge hacia la media poblacional teórica.

```

```python
# Teorema del Límite Central (TLC)
# Para muestras grandes (n ≥ 30), la distribución de las medias muestrales se aproxima a una normal, incluso si la población no es normal.
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
poblacion = np.random.exponential(scale=2, size=100000)

medias = []
for _ in range(1000):
    muestra = np.random.choice(poblacion, size=30)
    medias.append(np.mean(muestra))

plt.hist(medias, bins=30, density=True, color='skyblue')
plt.title("Teorema del Límite Central - Muestra de Exponencial")
plt.xlabel("Media muestral")
plt.ylabel("Frecuencia")
plt.show()
# Es un histograma de medias muestrales.
# Las muestras se generaron a partir de una distribución exponencial, que es altamente asimétrica (no normal).
# A pesar de ello, al tomar muchas muestras (por ejemplo, 1,000 muestras de tamaño n = 30 o más) y calcular
# sus medias, la distribución de esas medias muestrales se aproxima a una distribución normal.
# Eje X: Valores de las medias muestrales.
# Eje Y: Frecuencia (cuántas veces se observó cada media).
# El TLC establece que, si se toman muchas muestras aleatorias de una población con cualquier tipo de distribución
 # (incluso si es sesgada como la exponencial), la distribución de las medias muestrales se aproximará a una distribución
 # normal cuando el tamaño de muestra sea suficientemente grande (usualmente n ≥ 30).
```

```python
# Distribución Muestral de Proporciones
# Cuando estimamos proporciones en muestras grandes:
from scipy.stats import norm

p = 0.4 # Proporción poblacional de fumadores
p_hat = 0.45 # Proporción muestral
n = 100

se = np.sqrt(p * (1 - p) / n)
z = (p_hat - p) / se
prob = 1 - norm.cdf(z)

print(f"P(𝑝̂ > 0.45) = {prob:.4f}")
```

```python
# Ley de los grandes números
import numpy as np
import matplotlib.pyplot as plt

# Parámetros
mu = 0.5  # Media poblacional (probabilidad de cara)
n_experimentos = 10000  # Número de lanzamientos de moneda

# Simular lanzamientos y calcular la proporción acumulada de caras
muestras_acumuladas = np.cumsum(np.random.rand(n_experimentos) < mu) / np.arange(1, n_experimentos + 1)

# Gráfico
plt.figure(figsize=(10, 6))
plt.plot(range(1, n_experimentos + 1), muestras_acumuladas, label="Proporción acumulada de caras", color="blue")
plt.axhline(y=mu, color="red", linestyle="--", label=f"Media poblacional (μ={mu})")
plt.xlabel("Número de lanzamientos")
plt.ylabel("Proporción de caras")
plt.title("Ley de los Grandes Números: Convergencia hacia μ")
plt.legend()
plt.grid()
plt.show()
```

```python
# Teorema del Límite Central (TLC)
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon, norm

# Parámetros
n = 30 # Tamaño de la muestra
num_simulaciones = 10000 # Número de simulaciones
poblacion = expon.rvs(scale=1, size=num_simulaciones * n) # Distribución exponencial (no normal)

# Simular medias muestrales
# Tomamos 'num_simulaciones' muestras de tamaño 'n' de la población y calculamos sus medias
medias_muestrales = [np.mean(poblacion[i*n:(i+1)*n]) for i in range(num_simulaciones)]

# Gráfico
plt.figure(figsize=(10, 6))
plt.hist(medias_muestrales, bins=30, density=True, color="skyblue", edgecolor="black", label="Distribución de medias muestrales")

# Ajustar y graficar una distribución normal sobre las medias muestrales
x = np.linspace(min(medias_muestrales), max(medias_muestrales), 100)
plt.plot(x, norm.pdf(x, loc=np.mean(medias_muestrales), scale=np.std(medias_muestrales)), color="red", label="Distribución normal ajustada")

plt.title("Teorema del Límite Central: Convergencia hacia la normalidad")
plt.xlabel("Media muestral")
plt.ylabel("Densidad")
plt.legend()
plt.grid()
plt.show()
```

```python
from scipy.stats import norm
import numpy as np

# Datos
mu = 50  # Media poblacional
sigma = 10  # Desviación estándar poblacional
n = 36  # Tamaño de la muestra
x_bar = 52  # Media muestral deseada

# Calcular el error estándar
error_estandar = sigma / np.sqrt(n)

# Calcular el valor Z
z = (x_bar - mu) / error_estandar

# Calcular la probabilidad P(X_bar > 52) = P(Z > z_valor)
prob = 1 - norm.cdf(z)

print(f"Error estándar: {error_estandar:.2f}")
print(f"Valor Z: {z:.2f}")
print(f"P(X_bar > {x_bar}) = {prob:.4f} ({prob*100:.2f}%)")
```

```python
# Distribución muestral de proporciones
from scipy.stats import norm
import math

# Parámetros
p = 0.40  # Proporción poblacional
n = 100  # Tamaño de la muestra
p_hat = 0.45  # Proporción muestral deseada

# Calcular el error estándar (SE)
error_estandar = math.sqrt((p * (1 - p)) / n)

# Estandarizar a Z
z = (p_hat - p) / error_estandar

# Calcular la probabilidad P(p_hat > 0.45) = P(Z > z_valor)
prob = 1 - norm.cdf(z)

print(f"Error estándar: {error_estandar:.2f}")
print(f"Valor Z: {z:.2f}")
print(f"P(p_hat > {p_hat}) = {prob:.4f} ({prob*100:.2f}%)")
```

```python
# TLC
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Parámetros
mu = 10  # media de la distribución exponencial
n = 40   # tamaño de muestra
x_bar = 12  # valor objetivo para la media muestral

# --------- Método 1: Teorema del Límite Central (normal) ---------
sigma = mu  # Desviación estándar de una exponencial
se = sigma / np.sqrt(n)  # error estándar

# Calculamos la probabilidad de que la media muestral > 12
z = (x_bar - mu) / se
p_normal = 1 - stats.norm.cdf(z)

print(f"Usando el TLC (normal):")
print(f"Z = {z:.2f}")
print(f"P(media > 12) ≈ {p_normal:.4f}")

# --------- Método 2: Simulación directa (exponencial) ---------
simulaciones = 100000
medias_muestrales = np.mean(np.random.exponential(scale=mu, size=(simulaciones, n)), axis=1)
p_exponencial = np.mean(medias_muestrales > x_bar)

print("\nUsando simulación (distribución real):")
print(f"P(media > 12) ≈ {p_exponencial:.4f}")

# --------- Visualización ---------
plt.hist(medias_muestrales, bins=50, density=True, color='skyblue', alpha=0.7)
plt.axvline(x_bar, color='red', linestyle='--', label='x̄ = 12')
plt.title("Distribución de medias muestrales (Simulación)")
plt.xlabel("Media muestral")
plt.ylabel("Densidad")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

# Parámetros
a, b = 0, 20  # límites de la distribución uniforme
n = 30  # tamaño de muestra
simulaciones = 10000

# Simulación de medias muestrales
np.random.seed(0)
muestras = np.random.uniform(low=a, high=b, size=(simulaciones, n))
medias_muestrales = muestras.mean(axis=1)

# Parámetros teóricos para la normal
mu_teorica = (a + b) / 2
sigma_teorica = np.sqrt(((b - a) ** 2) / 12 / n)

# Gráfico
plt.figure(figsize=(8,5))
sns.histplot(medias_muestrales, kde=True, bins=40, color='skyblue', stat="density", label="Medias simuladas")
x = np.linspace(min(medias_muestrales), max(medias_muestrales), 200)
plt.plot(x, norm.pdf(x, mu_teorica, sigma_teorica), 'r--', label="Normal teórica (TLC)")
plt.title("Teorema del Límite Central - Distribución Uniforme U(0,20)")
plt.xlabel("Media muestral")
plt.ylabel("Densidad")
plt.axvline(mu_teorica, color='black', linestyle='--', label='Media poblacional')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Creamos una población Gamma (asimétrica positiva)
np.random.seed(42)
shape = 2     # parámetro de forma
scale = 3     # parámetro de escala
poblacion = np.random.gamma(shape=shape, scale=scale, size=10000)  # media = shape * scale = 6

# 2. Parámetros de muestreo
n = 40                # tamaño de cada muestra
num_muestras = 1000   # número de muestras

# 3. Generamos medias muestrales
medias_muestrales = [np.mean(np.random.choice(poblacion, size=n, replace=True)) for _ in range(num_muestras)]

# 4. Visualización
plt.figure(figsize=(12, 5))

# Distribución original Gamma
plt.subplot(1, 2, 1)
sns.histplot(poblacion, bins=30, kde=True, color='lightcoral')
plt.title("Distribución de la población (Gamma)")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")

# Distribución de las medias muestrales
plt.subplot(1, 2, 2)
sns.histplot(medias_muestrales, bins=30, kde=True, color='cornflowerblue')
plt.title(f"Distribución de medias muestrales (n={n})")
plt.xlabel("Media muestral")
plt.ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

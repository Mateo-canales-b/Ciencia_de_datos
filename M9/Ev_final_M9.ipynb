# EVALUACIÓN FINAL: ANÁLISIS DE MOVIMIENTOS MIGRATORIOS CON SPARK

Eres parte de un equipo de analistas encargado de estudiar tendencias de **migración humana** usando **Big Data**. Trabajarás con un dataset que contiene información sobre migraciones entre países, causas y **impacto socioeconómico** en origen y destino.


## Objetivos de la actividad
1. Aplicar conceptos de **Apache Spark / PySpark**.  
2. Explorar y transformar datos con **RDDs** y **DataFrames**.  
3. Ejecutar **consultas** con **Spark SQL**.  
4. Implementar **MLlib** para tareas predictivas.


## 1. Carga y exploración de datos (2 puntos)
- Carga el dataset en **Spark**.  
- Convierte los datos en **RDD** y en **DataFrame**.  
- Explora: muestra primeras filas, **esquema** y estadísticas descriptivas.


## 2. Procesamiento con RDDs y DataFrames (3 puntos)
- Transformaciones RDD: `filter`, `map`, `flatMap`.  
- Acciones RDD: `collect`, `take`, `count`.  
- Operaciones con DataFrames: **filtrado, agregaciones y ordenamiento**.  
- Escribe resultados en **formato Parquet**.


## 3. Consultas con Spark SQL (2 puntos)
- Registra el DataFrame como **tabla temporal**.  
- Consulta **principales países** de origen y destino.  
- Analiza **razones de migración** por región.


## 4. Predicción con MLlib (3 puntos)
- Prepara *features/label* para MLlib.  
- Aplica un modelo (p. ej., **regresión logística**) para predecir probabilidad de migración basada en factores socioeconómicos.  
- **Evalúa** el modelo y analiza su **precisión**.

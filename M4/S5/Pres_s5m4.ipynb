```python
# Módulo 4 sesión 5
# 🔹 Paso 1: Importar librerías necesarias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuramos el estilo visual
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
# sns.set(style="whitegrid")
# Esta línea configura el estilo estético por defecto para los gráficos de Seaborn.
# "whitegrid" agrega un fondo blanco con líneas de rejilla grises, lo cual:
# Mejora la legibilidad de los gráficos, especialmente para datos continuos.
# plt.rcParams["figure.figsize"] = (10, 6)
# Esto configura el tamaño predeterminado de todas las figuras generadas por Matplotlib.
# (10, 6) representa un gráfico de 10 pulgadas de ancho por 6 de alto.
# Esto evita que los gráficos se vean pequeños o mal proporcionados, especialmente al trabajar en entornos como Google Colab o Jupyter Notebook.
```

```python
# 🔹 Paso 2: Crear un DataFrame de ejemplo
np.random.seed(42)
df = pd.DataFrame({
    'edad': np.random.randint(18, 60, size=100),
    'ingresos': np.random.normal(3000, 500, size=100),
    'educacion': np.random.randint(6, 20, size=100),
    'sueño': np.random.normal(7, 1, size=100),
    'genero': np.random.choice(['Masculino', 'Femenino'], size=100)
})
print(df)

# Genera 100 valores enteros aleatorios entre 18 y 59 (el 60 no se incluye).
```

```python
# Gráficos de distribución: histplot() y kdeplot()
# Histograma con histplot() + curva de densidad
sns.histplot(data=df, x='ingresos', bins=20, kde=True, color='skyblue')
plt.title("Distribución de ingresos")
plt.show()
# sns.histplot() es una función de Seaborn que crea un histograma, una de las formas más comunes de visualizar la distribución de datos.
# bins=20: Divide los datos en 20 grupos (intervalos o "barras").
# kde=True: Agrega una curva de densidad sobre el histograma. Esto suaviza la forma de la distribución, como si fuera una "campana de Gauss".
# color='skyblue': Cambia el color de las barras a celeste.
```

```python
# Solo curva KDE
sns.kdeplot(data=df['sueño'], fill=True, color='purple')
plt.title("Curva de densidad - horas de sueño")
plt.show()
# sns.kdeplot() es una función de Seaborn que genera una estimación de la densidad de probabilidad
 # (KDE, por sus siglas en inglés: Kernel Density Estimation).
 # data=df['sueño']: Se le pasa la columna 'sueño' del DataFrame df, que contiene valores numéricos (por ejemplo, horas de sueño).
# fill=True: Rellena el área bajo la curva para que se vea como una montaña (más visual).
# color='purple': Define el color de la curva y el relleno.
```

```python
# Gráficos de dispersión y correlación: jointplot() y pairplot()
# Gráfico conjunto (dispersión + histogramas)
sns.jointplot(data=df, x='edad', y='ingresos', kind='scatter', color='green')
plt.show()
# kind='scatter': especifica que el gráfico central será un diagrama de dispersión (puntos).
# En el centro, un diagrama de dispersión que muestra cómo se relacionan edad e ingresos.
# En los bordes superior e izquierdo, aparecen histogramas que muestran cómo se distribuyen individualmente las variables edad e ingresos.

```

```python
# Pairplot: relaciones entre múltiples variables
sns.pairplot(df[['edad', 'ingresos', 'educacion']])
plt.show()
```

```python
# Gráfico de regresión: regplot()
# Línea de regresión entre educación e ingresos
sns.regplot(data=df, x='educacion', y='ingresos', scatter_kws={"color": "blue"}, line_kws={"color": "red"})
plt.title("Relación entre educación e ingresos")
plt.show()
```

```python
# Gráficos de variables categóricas: barplot(), countplot(), boxplot(), violinplot()
# Barplot: promedio de ingresos por género
sns.barplot(data=df, x='genero', y='ingresos')
plt.title("Promedio de ingresos por género")
plt.show()
```

```python
# Countplot: conteo de personas por género
sns.countplot(data=df, x='genero')
plt.title("Cantidad de observaciones por género")
plt.show()
```

```python
# Boxplot: distribución de ingresos por género
sns.boxplot(data=df, x='genero', y='ingresos')
plt.title("Boxplot de ingresos por género")
plt.show()
```

```python
# Violinplot: distribución + densidad de ingresos por género
sns.violinplot(data=df, x='genero', y='ingresos')
plt.title("Violinplot de ingresos por género")
plt.show()
# Este fragmento genera un gráfico de violín (violinplot) que permite visualizar cómo
# se distribuyen los ingresos según el género de las personas del DataFrame df.
# sns: se refiere a Seaborn, una biblioteca de visualización estadística.
# violinplot: es una función que combina:
# Un boxplot (diagrama de caja) que muestra la mediana y cuartiles.
# Una curva de densidad a cada lado, que da forma de violín, mostrando la distribución de la variable.
# Forma del violín:
# El ancho del violín en cada punto del eje Y indica cuántos datos hay con ese nivel de ingreso.
# La línea blanca horizontal en el centro del violín muestra la mediana.
# La caja negra vertical representa el rango intercuartílico (IQR).
# Las puntas indican el mínimo y máximo (sin considerar outliers extremos si los hubiera).
# Femenino:
# La distribución de ingresos es más simétrica, con un leve sesgo a la izquierda.
# La mayoría de los ingresos están concentrados entre los 2800 y 3400.
# Hay una mediana cercana a los 3000.
# Hay más personas con ingresos medios que bajos o muy altos.
#🔹 Masculino:
# La distribución es un poco más bimodal o más extendida (hay dos zonas anchas).
# Se observan dos grupos densos: uno cerca de 2600–2800 y otro cerca de 3300–3500.
# La mediana también está cerca de los 3000, pero con más dispersión en los extremos.
```

```python
# Gráfico de matriz: heatmap()
# Matriz de correlación
correlacion = df[['edad', 'ingresos', 'educacion', 'sueño']].corr()

# Heatmap con anotaciones
sns.heatmap(correlacion, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Heatmap de correlación entre variables numéricas")
plt.show()
```

sns.heatmap(...):
Es una función de Seaborn que genera un mapa de calor, ideal para matrices de correlación.
annot=True:
Muestra el valor numérico dentro de cada celda del heatmap.

cmap='coolwarm':
Es la paleta de colores.

Los valores negativos se ven en tonos azules.

Los valores positivos se ven en tonos rojos.

El color intermedio representa correlación cercana a 0.

fmt='.2f':
Formato de los números en las celdas → 2 decimales.

linewidths=0.5:
Agrega una línea divisoria entre las celdas del heatmap, de grosor 0.5.

```python
# Grillas de gráficos: PairGrid() y FacetGrid()
# PairGrid personalizado
g = sns.PairGrid(df[['edad', 'ingresos', 'educacion']], diag_sharey=False)
g.map_upper(sns.scatterplot)
g.map_diag(sns.histplot)
g.map_lower(sns.kdeplot, fill=True)
plt.suptitle("Grilla personalizada con PairGrid", y=1.02)
plt.show()
# Crea una grilla (g) usando las columnas edad, ingresos y educacion del DataFrame df.
# diag_sharey=False evita que los ejes Y en la diagonal tengan la misma escala.
# g.map_upper(sns.scatterplot) Dibuja diagramas de dispersión (scatter plots) en la parte superior de la grilla (por encima de la diagonal).
# Muestra cómo se relacionan las variables entre sí (correlación).
# g.map_diag(sns.histplot) En la diagonal principal dibuja histogramas, mostrando la distribución de cada variable de forma individual.
# g.map_lower(sns.kdeplot, fill=True)
# En la parte inferior de la grilla coloca curvas de densidad bivariadas (KDE).
# Estas curvas muestran cómo se distribuyen dos variables juntas (una estimación suave de la densidad conjunta).
# plt.suptitle(...) y plt.show() Agrega un título general a la grilla (y=1.02 lo posiciona un poco más arriba).

```

PairGrid() te permite crear una grilla de gráficos para comparar todas las combinaciones posibles de un conjunto de variables. Es muy útil para analizar relaciones bivariadas (entre dos variables) y univariadas (una sola variable).
 ¿Qué es un gráfico de densidad bivariada (KDE 2D)?
Es una forma de visualizar cómo se agrupan dos variables al mismo tiempo, suavizando la nube de puntos como si fuera una montaña o una “isla de densidad”.
Piensa que estás viendo un mapa de calor topográfico donde:

El centro (zona más azul oscuro) representa la zona donde hay más puntos (datos).

Los contornos concéntricos marcan niveles de densidad decreciente.

Similar a las curvas de nivel en un mapa de montaña.

El color azul más intenso indica mayor concentración de datos.
El núcleo azul oscuro muestra la zona donde hay más personas con cierta edad e ingresos.

A medida que te alejas, los contornos (como anillos) te dicen que la densidad de datos disminuye.

¿Para qué sirve?
Identificar agrupamientos naturales de los datos sin depender de un histograma con cajas.

Evitar el “ruido” visual que a veces producen los diagramas de dispersión si hay muchos puntos superpuestos.

Comparar relaciones entre variables suavemente: si hay una diagonal en los contornos, puede sugerir correlación.

Ejemplo:
Si ves una forma ovalada inclinada de abajo a la izquierda hacia arriba a la derecha, hay una correlación positiva (más edad, más ingresos).

Si ves un núcleo redondo y sin dirección clara (como en tu gráfico), eso indica que no hay una relación clara entre edad e ingresos.

```python
# FacetGrid: comparar la distribución de horas de sueño por género
g2 = sns.FacetGrid(df, col="genero")
g2.map(sns.histplot, "sueño", kde=True)
g2.fig.suptitle("Distribución de sueño según género", y=1.05)
plt.show()
```

col="genero" indica que se quiere dividir los datos en columnas según el género: por ejemplo, una para "Masculino" y otra para "Femenino".

Cada columna será un subgráfico independiente.

2. g2.map(sns.histplot, "sueño", kde=True)
Usa .map() para aplicar el gráfico que deseas: en este caso, sns.histplot (histograma).

"sueño" es la variable que se va a graficar.

kde=True añade una curva de densidad suavizada (KDE) encima del histograma, mostrando cómo se distribuyen las horas de sueño en cada género.

3. g2.fig.suptitle(...)
Agrega un título general para toda la grilla de gráficos.

y=1.05 ajusta la altura del título para que no se superponga a los subgráficos.



```python
# Módulo 4 sesión 5 Ejemplo 2
# 1. Importación de Librerías
# Es fundamental importar las librerías necesarias al inicio del script.
# Seaborn para la visualización estadística.
# Matplotlib.pyplot para personalizaciones adicionales de los gráficos.
# Pandas para la manipulación y creación de DataFrames.
# NumPy para operaciones numéricas y generación de datos de ejemplo.
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

print("Librerías importadas correctamente: Seaborn, Matplotlib, Pandas, NumPy.")

# 2. Creación de un Conjunto de Datos de Ejemplo
# Para demostrar los gráficos, crearemos un DataFrame ficticio que simule datos de clientes.
# Este DataFrame tendrá variables numéricas (Edad, Ingresos, Horas_Ejercicio)
# y categóricas (Género, Nivel_Educacion).
np.random.seed(42) # Para reproducibilidad de los resultados aleatorios

data = {
    'Edad': np.random.randint(18, 65, 200),
    'Ingresos': np.random.normal(50000, 15000, 200),
    'Horas_Ejercicio': np.random.randint(0, 10, 200),
    'Género': np.random.choice(['Masculino', 'Femenino'], 200),
    'Nivel_Educacion': np.random.choice(['Básica', 'Media', 'Universitaria', 'Posgrado'], 200, p=[0.15, 0.35, 0.40, 0.10]),
    'Satisfaccion_Cliente': np.random.randint(1, 6, 200) # De 1 (Muy Insatisfecho) a 5 (Muy Satisfecho)
}
df = pd.DataFrame(data)
print(df)
```

```python
# Añadimos algunas correlaciones artificiales para que los gráficos tengan más sentido.
df['Ingresos'] = df['Ingresos'] + (df['Edad'] * 500)
df['Horas_Ejercicio'] = df['Horas_Ejercicio'] + (df['Satisfaccion_Cliente'] * 0.5)

print("\nDataFrame de ejemplo creado:")
print(df.head())
print(f"\nDimensiones del DataFrame: {df.shape}")
print(f"Columnas del DataFrame: {df.columns.tolist()}")
print(f"Tipos de datos:\n{df.dtypes}")
```

# 3. Características de Seaborn (Explicación Teórica - no requiere código de ejecución aquí)

```python
# 4. Tipos de Gráficos con Seaborn

print("\n--- Demostración de Tipos de Gráficos con Seaborn ---")

# 4.1. Gráficos de Distribución de Observaciones

# histplot(): Histograma para visualizar la distribución de una variable numérica.
# Es el sucesor de distplot para histogramas y es recomendado.
plt.figure(figsize=(10, 6))
sns.histplot(df['Edad'], bins=15, kde=True, color='skyblue')
plt.title('Distribución de Edades')
plt.xlabel('Edad')
plt.ylabel('Frecuencia')
plt.grid(axis='y', alpha=0.75)
plt.show()
print("\nGráfico: Histograma de la distribución de Edades (histplot).")
print("Muestra la frecuencia de cada grupo de edad, con una línea KDE para la densidad.")

# kdeplot(): Estimación de la Densidad del Kernel para visualizar la distribución de una variable.
plt.figure(figsize=(10, 6))
sns.kdeplot(df['Ingresos'], fill=True, color='purple')
plt.title('Estimación de Densidad de Ingresos')
plt.xlabel('Ingresos')
plt.ylabel('Densidad')
plt.grid(axis='y', alpha=0.75)
plt.show()
print("\nGráfico: Estimación de Densidad de Ingresos (kdeplot).")
print("Representa la distribución de ingresos de forma suavizada.")
```

```python
# 4.2. Gráficos de Dispersión y Correlación de Variables

# jointplot(): Combina un gráfico de dispersión con histogramas/KDE marginales.
# Útil para ver la relación entre dos variables y sus distribuciones individuales.
plt.figure(figsize=(10, 8))
sns.jointplot(x='Edad', y='Ingresos', data=df, kind='reg', color='green') # 'kind' puede ser 'scatter', 'kde', 'hist', 'reg'
plt.suptitle('Relación entre Edad e Ingresos', y=1.02) # y ajusta la posición del título
plt.show()
print("\nGráfico: Jointplot de Edad vs Ingresos.")
print("Muestra la dispersión entre edad e ingresos, y sus distribuciones marginales. 'kind=reg' añade una línea de regresión.")

# pairplot(): Genera una matriz de gráficos de dispersión para todas las combinaciones de variables numéricas.
# Muy útil para un análisis exploratorio rápido de correlaciones.
# 'hue' permite colorear los puntos según una variable categórica.
plt.figure(figsize=(12, 12))
sns.pairplot(df[['Edad', 'Ingresos', 'Horas_Ejercicio', 'Satisfaccion_Cliente']], hue='Satisfaccion_Cliente', palette='viridis', diag_kind='kde')
plt.suptitle('Pairplot de Variables Numéricas por Satisfacción del Cliente', y=1.02)
plt.show()
print("\nGráfico: Pairplot de variables numéricas.")
print("Proporciona una vista rápida de las relaciones entre pares de variables numéricas, coloreadas por la satisfacción del cliente.")

# regplot(): Visualiza la relación lineal entre dos variables con una línea de regresión.
plt.figure(figsize=(10, 6))
sns.regplot(x='Edad', y='Ingresos', data=df, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.title('Regresión Lineal entre Edad e Ingresos')
plt.xlabel('Edad')
plt.ylabel('Ingresos')
plt.show()
print("\nGráfico: Regplot de Edad vs Ingresos.")
print("Muestra la relación lineal entre dos variables y su línea de regresión.")
```

```python
# 4.3. Gráficos de Variables Categóricas

# barplot(): Compara la media de una variable numérica para diferentes categorías.
plt.figure(figsize=(10, 6))
sns.barplot(x='Nivel_Educacion', y='Ingresos', data=df, ci='sd', palette='Blues') # ci='sd' muestra la desviación estándar
plt.title('Ingresos Medios por Nivel Educativo')
plt.xlabel('Nivel de Educación')
plt.ylabel('Ingresos Medios')
plt.show()
print("\nGráfico: Barplot de Ingresos Medios por Nivel Educativo.")
print("Compara los ingresos promedio para cada categoría de nivel educativo.")

# countplot(): Muestra el número de observaciones en cada categoría de una variable categórica.
plt.figure(figsize=(8, 5))
sns.countplot(x='Género', data=df, palette='viridis')
plt.title('Conteo de Observaciones por Género')
plt.xlabel('Género')
plt.ylabel('Conteo')
plt.show()
print("\nGráfico: Countplot de Observaciones por Género.")
print("Muestra la cantidad de personas en cada categoría de género.")

# boxplot(): Muestra la distribución, asimetría y posibles valores atípicos (outliers).
plt.figure(figsize=(10, 6))
sns.boxplot(x='Nivel_Educacion', y='Horas_Ejercicio', data=df, palette='coolwarm')
plt.title('Distribución de Horas de Ejercicio por Nivel Educativo')
plt.xlabel('Nivel de Educación')
plt.ylabel('Horas de Ejercicio')
plt.show()
print("\nGráfico: Boxplot de Horas de Ejercicio por Nivel Educativo.")
print("Visualiza la mediana, cuartiles y outliers de las horas de ejercicio para cada nivel educativo.")

# violinplot(): Combina un boxplot con una estimación de densidad del kernel.
# Proporciona una visualización más rica de la distribución de los datos.
plt.figure(figsize=(10, 6))
sns.violinplot(x='Nivel_Educacion', y='Ingresos', data=df, inner='quartile', palette='plasma')
plt.title('Distribución de Ingresos por Nivel Educativo (Violin Plot)')
plt.xlabel('Nivel de Educación')
plt.ylabel('Ingresos')
plt.show()
print("\nGráfico: Violinplot de Ingresos por Nivel Educativo.")
print("Similar al boxplot pero muestra la densidad de la distribución, dando una idea más clara de dónde se concentran los datos.")
```

```python
# 4.4. Gráficos de Matrices: Heatmap

# **Actividad Práctica Guiada: Generar un Heatmap para Visualizar la Correlación entre Variables**

print("\n--- Actividad Práctica Guiada: Heatmap de Correlación ---")

# Paso 1: Importar Librerías (Ya hecho al inicio del script)

# Paso 2: Crear un Conjunto de Datos (Ya hecho con el DataFrame 'df')
# Para esta actividad, vamos a seleccionar solo las variables numéricas para el heatmap.
df_numeric = df[['Edad', 'Ingresos', 'Horas_Ejercicio', 'Satisfaccion_Cliente']]
print("\nVariables numéricas seleccionadas para el heatmap:")
print(df_numeric.head())

# Paso 3: Calcular la Matriz de Correlación
# La función .corr() de Pandas calcula la matriz de correlación de Pearson por defecto.
correlation_matrix = df_numeric.corr()
print("\nMatriz de Correlación calculada:")
print(correlation_matrix)

# Paso 4: Generar el Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
# annot=True: Muestra los valores de correlación en las celdas.
# cmap='coolwarm': Escala de colores que va de azul (negativo) a rojo (positivo), pasando por blanco (cero).
# fmt=".2f": Formatea los valores a dos decimales.
# linewidths=.5: Añade líneas entre las celdas para mayor claridad.
plt.title('Matriz de Correlación de Variables Numéricas')
plt.show()
print("\nGráfico: Heatmap de la Matriz de Correlación.")
print("Visualiza las correlaciones entre pares de variables numéricas.")

# Paso 5: Interpretación del Heatmap
print("\n--- Interpretación del Heatmap ---")
print("Ejes: Las etiquetas de los ejes horizontal y vertical representan las variables (Edad, Ingresos, Horas_Ejercicio, Satisfaccion_Cliente).")
print("Colores:")
print("  - Colores cercanos al rojo oscuro (ej: 1.00): Correlación positiva fuerte (las variables aumentan o disminuyen juntas).")
print("  - Colores cercanos al azul oscuro (ej: -0.10): Correlación negativa fuerte (una variable aumenta mientras la otra disminuye).")
print("  - Colores claros (cercanos a 0): Correlación débil o inexistente.")
print("Valores: Cada celda muestra el coeficiente de correlación entre las dos variables correspondientes.")

print("\nEjemplos de Interpretación de nuestro Heatmap:")
print(f"  - Correlación Edad vs Ingresos: {correlation_matrix.loc['Edad', 'Ingresos']:.2f} (correlación positiva fuerte, como se diseñó).")
print(f"  - Correlación Edad vs Horas_Ejercicio: {correlation_matrix.loc['Edad', 'Horas_Ejercicio']:.2f} (correlación débil).")
print(f"  - Correlación Satisfaccion_Cliente vs Horas_Ejercicio: {correlation_matrix.loc['Satisfaccion_Cliente', 'Horas_Ejercicio']:.2f} (correlación positiva moderada, como se diseñó).")

# Paso 6: Personalización del Heatmap (Demostración)
plt.figure(figsize=(12, 10)) # Ajusta el tamaño del gráfico
sns.heatmap(correlation_matrix,
            annot=True,
            cmap='viridis', # Cambiamos el mapa de colores
            fmt=".2f",
            linewidths=1,   # Mayor grosor de líneas
            linecolor='black', # Color de las líneas
            cbar_kws={'label': 'Coeficiente de Correlación'}) # Etiqueta de la barra de color
plt.title('Matriz de Correlación Personalizada de Variables Numéricas', fontsize=16) # Tamaño del título
plt.xticks(fontsize=10) # Tamaño de las etiquetas del eje X
plt.yticks(fontsize=10, rotation=0) # Tamaño de las etiquetas del eje Y y rotación
plt.show()
print("\nGráfico: Heatmap Personalizado.")
print("Demostración de cómo personalizar el heatmap (tamaño, mapa de colores, grosor de líneas, etiquetas).")
```

```python
# 4.5. Grillas de Gráficos: PairGrid y FacetGrid

# PairGrid: Permite crear una grilla de gráficos de dispersión con más control.
# Puedes especificar diferentes tipos de gráficos para la diagonal (distribución)
# y para las celdas fuera de la diagonal (relaciones).
print("\n--- Demostración de Grillas de Gráficos ---")
g = sns.PairGrid(df_numeric)
g.map_upper(sns.scatterplot) # Gráficos de dispersión en la parte superior
g.map_lower(sns.kdeplot, cmap='Reds_d') # KDE plots en la parte inferior
g.map_diag(sns.histplot, color='gray') # Histogramas en la diagonal
# Nota: PairGrid no tiene un método set_titles directo para los subtítulos de cada eje.
# El título general de la figura se puede establecer con plt.suptitle().
plt.suptitle('PairGrid de Variables Numéricas', y=1.02, fontsize=16)
plt.show()
print("\nGráfico: PairGrid de Variables Numéricas.")
print("Demuestra mayor control sobre los tipos de gráficos en la diagonal y las partes superior/inferior.")

# FacetGrid: Util para explorar relaciones en subgrupos de datos.
# Permite dividir un conjunto de datos en subconjuntos basados en una variable categórica
# y graficar cada subconjunto por separado.
g = sns.FacetGrid(df, col='Nivel_Educacion', hue='Género', col_wrap=2, height=4, aspect=1.2, palette='tab10')
g.map(sns.scatterplot, 'Edad', 'Ingresos', alpha=0.7)
g.add_legend(title='Género')
g.set_axis_labels('Edad', 'Ingresos')
g.set_titles(col_template='Nivel: {col_name}') # Aquí sí se usa set_titles con template
plt.suptitle('FacetGrid de Edad vs Ingresos por Nivel Educativo y Género', y=1.02, fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.98]) # Ajusta el layout para que el título no se superponga
plt.show()
print("\nGráfico: FacetGrid de Edad vs Ingresos por Nivel Educativo y Género.")
print("Muestra la relación entre Edad e Ingresos dividida por Nivel Educativo, con colores según el Género.")
print("Útil para comparar patrones dentro de diferentes subgrupos.")

print("\n--- Conclusión del Script ---")
print("Este script ha cubierto la mayoría de los gráficos mencionados en los PDFs, proporcionando ejemplos ejecutables y explicaciones detalladas.")
print("Seaborn es una herramienta poderosa para el análisis visual de datos, permitiendo crear gráficos complejos con facilidad.")
print("Recuerda que la clave es elegir el tipo de gráfico adecuado para el tipo de datos y la pregunta que quieres responder.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Creamos un DataFrame simulado de empleados
df = pd.DataFrame({
    'Edad': [25, 30, 22, 45, 41, 29, 33, 55, 39, 27],
    'Salario': [3000, 3500, 2800, 7000, 6500, 3200, 4000, 9000, 6200, 3100],
    'Departamento': ['Ventas', 'TI', 'Marketing', 'TI', 'RRHH', 'Ventas', 'TI', 'RRHH', 'Ventas', 'Marketing']
})

# 1. Análisis Univariado
print("Estadísticas Descriptivas:\n", df.describe())

# Gráfico Boxplot para detectar outliers en Salario
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['Salario'])
plt.title("Boxplot del Salario")
plt.show()
# Mediana del salario
# La línea vertical dentro de la caja representa la mediana del salario, que está cerca de 4000.
# Esto indica que el 50% de los empleados gana menos de 4000 y el otro 50% gana más.
# Rango intercuartílico (IQR)
# La caja representa el rango entre el 1er cuartil (Q1) y el 3er cuartil (Q3). RIQ=Q3-Q1
# Aproximadamente el 50% central de los salarios se encuentra entre ~3200 y ~6200.
# Extremos (bigotes)
# Las líneas horizontales a los lados (bigotes) se extienden hasta los valores mínimo y máximo sin
# considerar valores atípicos (en este caso, no se muestran outliers).
# El mínimo visible está cerca de 2800 y el máximo llega a 9000.
# Asimetría
# La caja indica una distribución ligeramente asimétrica positiva.
# Es decir, hay algunos salarios altos que elevan el valor máximo.
#  Asimetría positiva (sesgo a la derecha):
# La caja es más corta a la izquierda y el bigote derecho es más largo.
# La mayoría de los datos está en los valores más bajos, pero hay valores altos alejados (colas largas a la derecha).
# Ejemplo: ingresos donde la mayoría gana poco y unos pocos ganan mucho.
# No se observan outliers explícitos
# No se muestran puntos fuera de los bigotes, lo que indica que no hay valores atípicos extremos en
# esta muestra.
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar datos desde la URL
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)
# Visualizar las primeras filas
print("Primeras filas del dataset Titanic:")
print(df.head())
```

```python
# 5. Detectar valores nulos por columna
print("\nValores nulos por columna:")
print(df.isnull().sum())
```

```python
# 6. Imputar valores faltantes
# Imputar 'age' con la mediana
df["age"].fillna(df["age"].median(), inplace=True)

# Imputar 'embarked' y 'embark_town' con la moda
df["embarked"].fillna(df["embarked"].mode()[0], inplace=True)
df["embark_town"].fillna(df["embark_town"].mode()[0], inplace=True)

# Eliminar la columna 'deck' por tener muchos nulos
df.drop(columns=["deck"], inplace=True)

# Verificar que ya no hay nulos
print("\n¿Aún hay valores nulos?")
print(df.isnull().sum())
```

```python
# Análisis de Componentes Principales + K-Means clustering

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import seaborn as sns
import matplotlib.pyplot as plt

# Variables numéricas para clustering
features = ['Economy', 'Family', 'Health', 'Freedom', 'Generosity', 'Corruption', 'Dystopia', 'Job Satisfaction']
df_cluster = df[features].dropna()
# features: Esta lista define las columnas de tu DataFrame original (df) que se usarán para el análisis de agrupación.
# Son los "componentes de felicidad" como 'Economía', 'Familia', 'Salud', etc.
# df_cluster = df[features].dropna():

# Crea un nuevo DataFrame llamado df_cluster que contiene solo las columnas especificadas en features.

# .dropna() elimina cualquier fila que tenga valores faltantes (NaN) en cualquiera de estas columnas seleccionadas.
# Este es un paso crucial porque la mayoría de los algoritmos de aprendizaje automático no pueden manejar datos faltantes.

# Estandarización
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster)
# scaler = StandardScaler(): Crea una instancia de StandardScaler. La estandarización es un paso común de preprocesamiento
# en el aprendizaje automático, especialmente para algoritmos que son sensibles a la escala de las características de entrada
 # (como K-Means).

# X_scaled = scaler.fit_transform(df_cluster):
# fit_transform hace dos cosas:
# fit: Calcula la media y la desviación estándar para cada característica en df_cluster.
# transform: Luego usa estas medias y desviaciones estándar calculadas para transformar los datos. Para cada característica,
# resta la media y divide por la desviación estándar. Esto resulta en datos con una media de 0 y una desviación estándar de 1,
# lo que efectivamente pone todas las características en la misma escala. Esto evita que las características con valores más
# grandes (por ejemplo, 'Economía') dominen el proceso de agrupación sobre las características con valores más pequeños
 # (por ejemplo, 'Corrupción').




# PCA a 2 dimensiones
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# pca = PCA(n_components=2): Crea una instancia de PCA (Análisis de Componentes Principales). PCA es una técnica de reducción de
# dimensionalidad. Aquí, n_components=2 significa que queremos reducir las 8 características originales a 2

# X_pca = pca.fit_transform(X_scaled):

# fit_transform nuevamente hace dos cosas:

# fit: Aprende los componentes principales a partir de los datos X_scaled. Los componentes principales son variables nuevas y no
# correlacionadas que capturan la mayor parte de la varianza en los datos originales.

# transform: Proyecta los datos X_scaled en estos 2 componentes principales. X_pca será una matriz NumPy con dos columnas, que
# representan los valores de los dos primeros componentes principales para cada país. Esto hace que los datos sean mucho más
# fáciles de visualizar.


# KMeans con 4 clusters
kmeans = KMeans(n_clusters=4, random_state=42)
clusters = kmeans.fit_predict(X_pca)
# kmeans = KMeans(n_clusters=4, random_state=42):

# Crea una instancia del algoritmo de clustering KMeans.
# n_clusters=4: Le dice al algoritmo que agrupe los datos en 4 clusters distintos.
# random_state=42: Se usa para asegurar la reproducibilidad. K-Means implica cierta inicialización aleatoria, por lo que establecer
# un random_state asegura que obtendrás los mismos resultados cada vez que ejecutes el código con los mismos datos de entrada.
# clusters = kmeans.fit_predict(X_pca):

# fit_predict hace dos cosas:

# fit: Ejecuta el algoritmo K-Means sobre los datos X_pca. El algoritmo asigna iterativamente los puntos de datos al centroide del cluster
# más cercano y luego actualiza los centroides hasta que converge.

# predict: Devuelve una matriz (clusters) donde cada elemento es un entero (0, 1, 2 o 3 en este caso) que representa la asignación de
# cluster para cada país.


# Agregar resultados al DataFrame original
df_clusterizado = df.loc[df_cluster.index].copy()
df_clusterizado['PCA1'] = X_pca[:, 0]
df_clusterizado['PCA2'] = X_pca[:, 1]
df_clusterizado['Cluster'] = clusters
# df_clusterizado = df.loc[df_cluster.index].copy():
# Crea un nuevo DataFrame df_clusterizado seleccionando las filas del DataFrame original df que corresponden a las filas que se utilizaron
# para el clustering (es decir, aquellas sin valores faltantes). .copy() es importante para evitar advertencias de copia (SettingWithCopyWarning)
# más adelante.
# df_clusterizado['PCA1'] = X_pca[:, 0]: Agrega una nueva columna llamada 'PCA1' a df_clusterizado. Esta columna contiene los valores del primer
# componente principal para cada país.

# df_clusterizado['PCA2'] = X_pca[:, 1]: Agrega una nueva columna llamada 'PCA2' a df_clusterizado. Esta columna contiene los valores del segundo
# componente principal para cada país.

# df_clusterizado['Cluster'] = clusters: Agrega una nueva columna llamada 'Cluster' a df_clusterizado. Esta columna contiene la asignación de
# cluster (0, 1, 2 o 3) para cada país, según lo determinado por el algoritmo K-Means.


# Visualización
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df_clusterizado, x='PCA1', y='PCA2', hue='Cluster', palette='Set2', s=80)
plt.title('Clustering de países por componentes de felicidad (PCA + KMeans)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.grid(True)
plt.show()

# Mostrar tabla agrupada
df_clusterizado[['Country', 'Region', 'Happiness Score', 'Cluster']].sort_values('Cluster').head(10)

# plt.figure(figsize=(10, 6)): Crea una nueva figura para el gráfico con un tamaño especificado (10 pulgadas de ancho, 6 pulgadas de alto).

# sns.scatterplot(...): Esto utiliza la librería Seaborn para crear un gráfico de dispersión:

# data=df_clusterizado: Especifica el DataFrame que contiene los datos a graficar.

# x='PCA1': Mapea la columna 'PCA1' al eje X.

# y='PCA2': Mapea la columna 'PCA2' al eje Y.

# hue='Cluster': Esta es la parte clave para la visualización. Colorea los puntos de datos según su asignación de 'Cluster'. Esto te permite
# distinguir visualmente los diferentes grupos.

# palette='Set2': Establece la paleta de colores para los clusters.

# s=80: Establece el tamaño de los marcadores del gráfico de dispersión.

# plt.title(...), plt.xlabel(...), plt.ylabel(...): Establecen el título y las etiquetas de los ejes para el gráfico, haciéndolo más informativo.

# plt.grid(True): Agrega una cuadrícula al gráfico para una mejor legibilidad.

# plt.show(): Muestra el gráfico generado.

# En resumen, este código toma datos multidimensionales sobre países, los simplifica en dos dimensiones clave usando PCA, y luego agrupa a los
# países con perfiles de felicidad similares en clusters distintos usando K-Means, para finalmente mostrar estos grupos en un gráfico 2D.
```

```python
import pandas as pd
import zipfile
import requests
from io import BytesIO

# Descargar y descomprimir
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00222/bank.zip"
r = requests.get(url)
z = zipfile.ZipFile(BytesIO(r.content))
z.extractall()

# Cargar dataset principal
df = pd.read_csv("bank-full.csv", sep=";")
print(df.head())
print(df['y'].value_counts())
# El dataset que estás utilizando es el famoso "Bank Marketing Dataset", comúnmente usado en problemas de clasificación binaria.
# La variable objetivo y indica si el cliente aceptó (yes) o no (no) una oferta de producto bancario (como un depósito a plazo).
#  Variables del Dataset
# Variable	Tipo	Significado
# age	Numérica	Edad del cliente (en años).
# job	Categórica	Tipo de trabajo del cliente (ej. management, technician, blue-collar, etc.).
# marital	Categórica	Estado civil del cliente: single, married, divorced.
# education	Categórica	Nivel educativo: primary, secondary, tertiary o unknown.
# default	Binaria	Si el cliente tiene historial de crédito en mora (no o yes).
# balance	Numérica	Saldo promedio anual de la cuenta del cliente (en euros).
# housing	Binaria	Si el cliente tiene un préstamo hipotecario (yes o no).
# loan	Binaria	Si el cliente tiene un préstamo personal (yes o no).
# contact	Categórica	Tipo de contacto usado en la campaña: cellular, telephone, unknown.
# day	Numérica	Día del mes en el que se contactó al cliente.
# month	Categórica	Mes en el que se contactó al cliente (ej. may, jun, nov, etc.).
# duration	Numérica	Duración de la última llamada en segundos. (Muy importante: si no se contactó, el valor es 0).
# campaign	Numérica	Número de contactos realizados con este cliente durante esta campaña.
# pdays	Numérica	Días desde el último contacto del cliente en una campaña anterior. -1 significa que no fue contactado previamente.
# previous	Numérica	Número de contactos anteriores realizados antes de esta campaña.
# poutcome	Categórica	Resultado de la campaña anterior: success, failure, nonexistent, unknown.
# y	Binaria	Variable objetivo (target): ¿El cliente aceptó suscribirse al producto (depósito a plazo fijo)? yes o no.
```

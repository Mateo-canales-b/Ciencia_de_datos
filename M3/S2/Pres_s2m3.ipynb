```python
# Importar la librería pandas
import pandas as pd
```

```python
# Crear un DataFrame con datos simulados de empleados
data = {
    'Nombre': ['Ana', 'Luis', 'Marta', 'Carlos', 'Patricia', 'Jorge', 'Lucía', 'Pedro', 'Sofía', 'Francisco'],
    'Edad': [25, 34, 28, 45, 41, 30, 29, 38, 27, 32],
    'Departamento': ['Ventas', 'RRHH', 'Ventas', 'TI', 'TI', 'RRHH', 'Ventas', 'TI', 'Ventas', 'TI'],
    'Salario': [850, 920, 780, 1100, 1050, 900, 820, 1000, 790, 1020]
}
# Convertir el diccionario a un DataFrame
df = pd.DataFrame(data)
# Mostrar las primeras filas
print("Primeras filas del DataFrame:")
print(df.head())
```

```python
# Mostrar estadísticas descriptivas
print("\nEstadísticas descriptivas:")
print(df.describe())
```

```python
# Filtrar empleados mayores de 30 años
print("\nEmpleados mayores de 30 años:")
print(df[df['Edad'] > 30])
```

```python
# Calcular el salario promedio
salario_promedio = df['Salario'].mean()
print("\nSalario promedio:", salario_promedio)
```

```python
# Conteo por departamento
print("\nNúmero de empleados por departamento:")
print(df['Departamento'].value_counts())
```

```python
# Filtrar empleados del departamento de TI con salario mayor a 1000
print("\nEmpleados del departamento TI con salario mayor a 1000:")
print(df[(df['Departamento'] == 'TI') & (df['Salario'] > 1000)])
```

```python
# Actividad Avanzada - Análisis de Ventas con Pandas

import pandas as pd
import numpy as np

# 1. Crear un DataFrame con información de ventas
# Algunas celdas contienen valores nulos o erróneos para trabajar en limpieza de datos
data = {
    "Vendedor": ["Ana", "Luis", "Ana", "Pedro", "Luis", "Sofía", "Pedro", "Ana", "Sofía", None],
    "Región": ["Norte", "Centro", "Norte", "Sur", "Centro", "Norte", "Sur", "Norte", "Norte", "Centro"],
    "Ventas": [2500, 3000, np.nan, 1800, 3100, 4000, -500, 2800, 3900, 3200],
    "Mes": ["Enero", "Enero", "Febrero", "Febrero", "Febrero", "Marzo", "Marzo", "Marzo", "Marzo", "Enero"]
}

df = pd.DataFrame(data)

print("Data original con errores y valores faltantes:\n")
print(df)
```

```python
# 2. Eliminar filas con vendedor nulo
df = df.dropna(subset=["Vendedor"])
```

```python
# 3. Reemplazar ventas negativas o nulas con la media de ventas válidas
ventas_validas = df["Ventas"][(df["Ventas"] > 0) & (~df["Ventas"].isna())]
media_ventas = ventas_validas.mean()
df["Ventas"] = df["Ventas"].apply(lambda x: media_ventas if pd.isna(x) or x <= 0 else x)

print("\n Data limpia (sin nulos y sin valores negativos en ventas):\n")
print(df)

ventas_validas = df["Ventas"].dropna()           # Elimina valores nulos
ventas_validas = ventas_validas[ventas_validas > 0]  # Filtra solo los positivos
```

```python
# 4. Agregar una columna con comisión del 10%
df["Comisión"] = df["Ventas"] * 0.10
# Imprimir solo la columna Comisión
print(df["Comisión"])
```

```python
# 5. Mostrar las ventas totales por vendedor
ventas_por_vendedor = df.groupby("Vendedor")["Ventas"].sum()
print("\n Ventas totales por vendedor:\n")
print(ventas_por_vendedor)
```

```python
# 6. Mostrar promedio de ventas por región
promedio_por_region = df.groupby("Región")["Ventas"].mean()
print("\n Promedio de ventas por región:\n")
print(promedio_por_region)
```

```python
# 7. Crear una columna indicando si superó el objetivo (mayor a $3000)
df["Objetivo Cumplido"] = df["Ventas"] > 3000

print("\n Data final con comisiones y objetivos:\n")
print(df)
```

```python
# 8. Filtrar ventas de Marzo donde se cumplió el objetivo
print("\n Ventas de Marzo con objetivo cumplido:\n")
print(df[(df["Mes"] == "Marzo") & (df["Objetivo Cumplido"])])
```

Ejemplo 2: Filtrar datos con Pandas y aplicar operaciones NumPy

```python
import numpy as np
import pandas as pd

# DataFrame con edades
df = pd.DataFrame({"Nombre": ["Ana", "Luis", "Carlos", "Marta"],
                   "Edad": [25, 15, 42, 37]})
```

```python
# Filtrar personas mayores de 30
mayores = df[df["Edad"] > 30]
```

```python
# Aplicar raíz cuadrada a las edades filtradas (usando NumPy)
edades_sqrt = np.sqrt(mayores["Edad"])

print("Raíz cuadrada de edades mayores de 30:\n", edades_sqrt)
```

```python
# Ejemplo 3: Estadísticas combinadas
import numpy as np
import pandas as pd

# Datos de ingresos semanales
ingresos = np.array([200, 300, 250, 400, 100, 150])

# Crear DataFrame
df = pd.DataFrame({"Ingresos": ingresos})
```

```python
# Estadísticas con NumPy y Pandas
media = np.mean(df["Ingresos"])        # NumPy
mediana = df["Ingresos"].median()      # Pandas
desv_std = np.std(df["Ingresos"])      # NumPy

print(f"Media: {media}, Mediana: {mediana}, Desviación estándar: {desv_std}")
```

Importancia: Puedes usar indistintamente métodos de NumPy o Pandas según te convenga; combinarlos agiliza el análisis y visualización.

Ejemplo: Diferencia entre np.array y pd.Series

```python
import numpy as np
import pandas as pd

# NumPy array
edades_array = np.array([25, 30, 35])
print("Array de NumPy:")
print(edades_array)
print("Índices de NumPy:", list(range(len(edades_array))))
```

```python
# Serie de Pandas con índices personalizados
edades_serie = pd.Series([25, 30, 35], index=["Ana", "Luis", "Marta"])
print("\nSerie de Pandas:")
print(edades_serie)
print("Índices de la Serie:", edades_serie.index)
```

Característica	NumPy (np.array)
Índices	Siempre numéricos (0, 1, 2, ...)
Acceso a elementos	Por posición
Funcionalidad extra	Cálculo numérico rápido

Pandas (pd.Series)
Personalizables (pueden ser strings, fechas)
Por posición y por etiqueta
Etiquetado + más opciones analíticas

```python
print(edades_array[1])        # Accede al segundo valor (30)
print(edades_serie["Luis"])   # Accede al valor de 'Luis' (30)
```

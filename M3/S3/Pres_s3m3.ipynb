```python
# 3. Personalizar lectura CSV (separador y valores nulos)
files.upload()  # Subir archivo 'ejemplo_datos_personalizado.csv'
df_sep = pd.read_csv("ejemplo_datos_personalizado.csv", sep=';', na_values=['NA', 'N/A', ''])
print("\nCSV con separador ';' y nulos personalizados:")
print(df_sep.head())
```

```python
# 4. Leer archivo Excel con nulos personalizados
files.upload()  # Subir archivo 'ejemplo_datos_personalizado.xlsx'
df_excel = pd.read_excel("ejemplo_datos_personalizado.xlsx", na_values=["NA", "N/A", ""])
print("\nExcel con valores nulos personalizados:")
print(df_excel.head())
```

```python
# 5. Verificar la cantidad de datos y tipos de columnas
print("Resumen de df (ejemplo_datos.csv):")
print(df.info())

print("\nResumen de df_excel (ejemplo_datos.xlsx):")
print(df_excel.info())

print("\nResumen de df_sep (ejemplo_datos_personalizado.csv):")
print(df_sep.info())
```

```python
# 6. Estadísticas descriptivas
print("\nEstadísticas descriptivas df:")
print(df.describe())

print("\nEstadísticas descriptivas df_excel:")
print(df_excel.describe())

print("\nEstadísticas descriptivas df_sep:")
print(df_sep.describe())
```

```python
# 7. Limpiar y calcular: reemplazar ventas nulas o negativas por la media y agregar columna de comisión y utilidad
def limpiar_y_calcular(df):
    if "Ventas" in df.columns:
        ventas_validas = df["Ventas"][(df["Ventas"] > 0) & (~df["Ventas"].isna())]
        media = ventas_validas.mean()
        df["Ventas"] = df["Ventas"].apply(lambda x: media if pd.isna(x) or x <= 0 else x)
        df["Comisión"] = df["Ventas"] * 0.10
        df["Utilidad"] = df["Ventas"] - df.get("Costos", 0) - df.get("Costos_Operativos", 0)
    return df

df = limpiar_y_calcular(df)
df_excel = limpiar_y_calcular(df_excel)
df_sep = limpiar_y_calcular(df_sep)
```

```python
# 8. Mostrar algunos resultados
print("\nDatos procesados df:")
print(df.head())

print("\nDatos procesados df_excel:")
print(df_excel.head())

print("\nDatos procesados df_sep:")
print(df_sep.head())
```

```python
# Exportar a archivos CSV
df.to_csv("df_limpio.csv", index=False)
df_excel.to_csv("df_excel_limpio.csv", index=False)
df_sep.to_csv("df_sep_limpio.csv", index=False)
```

Instrucciones:
1. Importa las librerías necesarias (pandas).

2. Carga los archivos respetando:

Para el CSV: Usa sep=';' y considera como nulos: "NA", "-" y "".

Para el Excel: considera como nulos: "NA", "-" y "".

3. Muestra las primeras filas de ambos DataFrames.

4. Cuenta cuántos datos nulos hay por columna en ambos archivos.

5. Llena los valores nulos en "Precio" con la media de los precios válidos.

6. Llena los valores nulos en "Ciudad" con "Desconocido".

7. Crea una nueva columna llamada Ingreso Total en el archivo Excel calculando Unidades * Precio (une ambas tablas por ID).

8. Filtra las filas donde el ingreso total sea mayor a 3000 y guarda esa tabla como ventas_altas.xlsx.

9. Exporta la tabla de productos corregida a un nuevo archivo llamado productos_limpios.csv.

```python
# 1.
import pandas as pd
```

```python
# 2. Cargar archivos y mostrar primeras filas
from google.colab import files

# Subir CSV
uploaded = files.upload()  # productos.csv
df_productos = pd.read_csv("productos.csv", sep=';', na_values=["NA", "-", ""])
print("Primeras filas de productos.csv:")
print(df_productos.head())
```

```python
# Subir Excel
uploaded = files.upload()  # ventas_tienda.xlsx
df_ventas = pd.read_excel("ventas_tienda.xlsx", na_values=["NA", "-", ""])
print("\nPrimeras filas de ventas_tienda.xlsx:")
print(df_ventas.head())
```

```python
# 3. Contar valores nulos
print("\nNulos en productos.csv:")
print(df_productos.isna().sum())

print("\nNulos en ventas_tienda.xlsx:")
print(df_ventas.isna().sum())
```

```python
# 4. Rellenar nulos
# Rellenar nulos en Precio con la media
media_precio = df_productos["Precio"].mean()
df_productos["Precio"] = df_productos["Precio"].fillna(media_precio)

# Rellenar nulos en Ciudad con "Desconocido"
df_ventas["Ciudad"] = df_ventas["Ciudad"].fillna("Desconocido")

# Rellenar nulos en Unidades con 0
df_ventas["Unidades"] = df_ventas["Unidades"].fillna(0)
```

```python
# Paso 5: Unir ambas tablas por ID y calcular "Ingreso Total"
# Unir df_ventas con df_productos según el ID de producto
df_ventas_completo = pd.merge(df_ventas, df_productos, how='left', left_on="ID_Producto", right_on="ID")

# Calcular el Ingreso Total: Unidades * Precio
df_ventas_completo["Ingreso_Total"] = df_ventas_completo["Unidades"] * df_ventas_completo["Precio"]
```

```python
# Paso 6: Filtrar ingresos mayores a 3000 y guardar en Excel
# Filtrar ventas con ingreso total mayor a 3000
ventas_altas = df_ventas_completo[df_ventas_completo["Ingreso_Total"] > 3000]

# Exportar a archivo Excel
ventas_altas.to_excel("ventas_altas.xlsx", index=False)
```

```python
# Paso 7: Exportar tabla de productos corregida a CSV
# Exportar la tabla de productos corregida
df_productos.to_csv("productos_limpios.csv", index=False, sep=";")
# Ver la tabla en pantalla
print(df_productos.head())  # Muestra las primeras filas
```

```python
# Ver los primeros 5 países con mayor población proyectada
# Paso 1: Renombrar columnas para facilitar el trabajo
df_web.columns = [
    "Ranking",
    "Pais",
    "Poblacion_2025",
    "Porcentaje_Mundial",
    "Cambio_Anual_Porc",
    "Cambio_Anual_Absoluto",
    "Cambio_Absoluto_Porc",
    "Anos_Duplicacion",
    "Censo_Ultimo",
    "Fecha_Censo",
    "Tipo",
    "Enlace"
]

# Paso 2: Limpiar la columna de población (quitar espacios y caracteres invisibles)
df_web["Poblacion_2025"] = df_web["Poblacion_2025"].astype(str)                # Asegura tipo texto
df_web["Poblacion_2025"] = df_web["Poblacion_2025"].str.replace(r"[^\d]", "", regex=True)  # Deja solo números
#astype(str)
# Convierte todos los valores a cadena de texto.
# Esto es importante para luego aplicar métodos .str.
# .str es un acceso especial en Pandas que te permite aplicar métodos de cadenas (strings) a toda una columna de un DataFrame.
# Solo puedes usar .str en columnas de tipo texto (object o string).

# .str.replace(r"[^\d]", "", regex=True)
# Usa una expresión regular para eliminar todo lo que no sea un número (0–9).

# [^\d] = “todo carácter que no sea un dígito”

# "" = lo reemplaza con una cadena vacía

#Resultado: elimina espacios, puntos, comas, caracteres invisibles, etc.
# Ejemplo:
# Antes (texto original)	Después de limpieza
# "1 417 492 000"	            "1417492000"
# "342 181 000"	              "342181000"
# "–" (guion largo)         	"" (vacío)


# Paso 3: Convertir a número, errores a NaN
df_web["Poblacion_2025"] = pd.to_numeric(df_web["Poblacion_2025"], errors='coerce')
# Convierte la columna limpia a valores numéricos (int64 o float64).
# Si alguna celda no puede convertirse (por ejemplo, queda vacía o con caracteres extraños), se convierte en NaN (valor nulo).
# errors='coerce' = convierte errores a NaN en lugar de lanzar un error.

# Paso 4: Ordenar y mostrar los 5 primeros
df_top = df_web.sort_values("Poblacion_2025", ascending=False)
print(df_top[["Pais", "Poblacion_2025"]].head())
# Ordena el DataFrame df_web según la columna "Poblacion_2025".
# El parámetro ascending=False indica que el orden será descendente, es decir, de mayor a menor población.
# El resultado se guarda en df_top.

# Selecciona solo las columnas "Pais" y "Poblacion_2025" del DataFrame ordenado.
# Muestra las primeras 5 filas con .head() → que ahora son los 5 países con mayor población proyectada para 2025.
```

```python
# Contar cuántos países tienen datos censales reales
# Contar por tipo
conteo_tipos = df_web["Tipo"].value_counts()
print(conteo_tipos)
# df_web["Tipo"] accede a la columna "Tipo", que indica el tipo de fuente del dato poblacional.
# .value_counts() cuenta cuántas veces aparece cada tipo distinto.
# El resultado es una serie de Pandas con la frecuencia de cada tipo.
```

Cada letra en la columna "Tipo" representa el tipo de fuente o método usado para obtener la cifra de población. Aunque Wikipedia no siempre explica todos en detalle, estos suelen ser:

E: Estimación

C: Censo

O: Otra fuente oficial

A: Reloj de población automatizado

M: Modelado estadístico

F: Fuente oficial sin especificar tipo

T: Temporal o transitorio

R: Revisión o reproyección

P: Proyección futura

B: Base de datos u organismo especial

Tipo: Esta probablemente es una fila de encabezado mal interpretada como dato → puedes eliminarla.

```python
# Eliminar la fila errónea
df_web = df_web[df_web["Tipo"] != "Tipo"]

# Verifica nuevamente los tipos
conteo_tipos = df_web["Tipo"].value_counts()
print(conteo_tipos)
```

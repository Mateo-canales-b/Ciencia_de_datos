Aprendizaje Supervisado
🔹 Definición: El modelo aprende con ayuda de etiquetas. Se le dicen las respuestas correctas durante el entrenamiento.

Ejemplo:
Tienes una tabla con características y etiquetas, así:

Color	Tamaño	Forma	Fruta
Rojo	Mediano	Redonda	Manzana 🍎
Amarillo	Grande	Alargada	Plátano 🍌
Morado	Pequeño	Redonda	Uva 🍇

Tu tarea es predecir el tipo de fruta, dado su color, tamaño y forma. Ya tienes las respuestas (etiquetas), así que entrenas un modelo (por ejemplo, un árbol de decisión) para aprender esa relación.

Cómo explicarlo:

“El modelo sabe qué fruta es cada una mientras aprende. Luego, cuando vea una fruta nueva, intentará decirnos cuál es basándose en lo que aprendió”.

Ejemplos reales de aprendizaje supervisado:

Clasificar correos como spam o no spam.

Diagnosticar una enfermedad (usando historiales de pacientes).

Predecir el precio de una casa.


Aprendizaje No Supervisado
🔹 Definición: El modelo no tiene etiquetas. Solo se le da la información, y debe encontrar patrones por sí mismo.

Ejemplo:
Ahora solo tienes esto:

Color	Tamaño	Forma
Rojo	Mediano	Redonda
Amarillo	Grande	Alargada
Morado	Pequeño	Redonda

Tú no le dices al modelo qué fruta es cuál. El modelo agrupa las frutas en base a sus similitudes, por ejemplo:

Grupo 1: rojo, redondo, mediano → probablemente manzanas.

Grupo 2: amarillo, grande, alargado → probablemente plátanos.

Grupo 3: morado, pequeño, redondo → probablemente uvas.

Cómo explicarlo:

“El modelo no sabe qué fruta es cuál. Solo trata de agrupar frutas parecidas, como cuando ves muchas cosas juntas y las organizas por color o forma, aunque no sepas sus nombres”.

Ejemplos reales de aprendizaje no supervisado:

Agrupar clientes según comportamiento de compra (marketing).

Detectar fraudes o anomalías.

Segmentar imágenes o documentos por similitud.

Resumen visual para tus estudiantes:
Característica	Supervisado 🧭	No Supervisado 🔍
¿Tiene etiquetas?	✅ Sí	❌ No
¿Se sabe la respuesta?	✅ Sí (durante el entrenamiento)	❌ No
Ejemplo clásico	Clasificar frutas 🍎🍌🍇	Agrupar frutas similares 🍏🍌🍇
¿Cuál es la tarea?	Clasificar / predecir	Agrupar / descubrir patrones

```python
# Clase 07/08/25 Módulo 7 sesión 1
# Ejemplo 1: Clusterización con K-Means
# Este código implementa el algoritmo
# K-Means, que es una de las técnicas de clusterización más populares. El objetivo es agrupar datos sintéticos de clientes en
# grupos (clústeres) basados en sus características, en este caso, "ingreso" y "gasto promedio mensual".
```

```python
# SECCIÓN: CLUSTERIZACIÓN CON K-MEANS
# Este código demuestra cómo usar el algoritmo K-Means para agrupar datos de clientes
# en clústeres basados en sus similitudes.
# El objetivo de la clusterización es dividir un conjunto de datos en grupos (clústeres) donde los datos dentro de un mismo grupo son más similares entre
# sí que con los de otros grupos
# ==============================================================================

# 1. Importar las librerías necesarias
# KMeans: El algoritmo de clustering K-Means.
# numpy: Para la manipulación de arrays.
# matplotlib.pyplot: Para visualizar los resultados.
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

print("--- DEMOSTRACIÓN DE CLUSTERIZACIÓN CON K-MEANS ---")

# 2. Crear datos de ejemplo
# Los datos representan a 5 clientes con 2 características: [ingreso, gasto promedio mensual].
# Creamos datos ficticios para el ejemplo.
datos_clientes = np.array([
    [30, 300],  # Cliente 1
    [40, 400],  # Cliente 2
    [25, 250],  # Cliente 3
    [80, 850],  # Cliente 4
    [50, 500],  # Cliente 5
    [85, 900],  # Cliente 6
    [20, 220],  # Cliente 7
    [55, 530],  # Cliente 8
])
# datos_clientes = np.array(...): Crea un array de NumPy llamado datos_clientes.
# Cada sub-array (por ejemplo, [30, 300]) representa a un cliente.
# Cada valor dentro de los sub-arrays es una característica del cliente: el primer valor es su ingreso y el segundo es su gasto promedio mensual.
# Se han creado datos ficticios para el ejemplo para simular un escenario real en el que podríamos querer agrupar clientes según su comportamiento.
# Esta parte del código está simulando un conjunto de datos de 8 clientes, donde cada cliente tiene dos variables numéricas que nos servirán de base
# para que el algoritmo de K-Means busque patrones y similitudes entre ellos. El objetivo es que K-Means descubra si estos clientes pueden agruparse
# naturalmente en diferentes segmentos (por ejemplo, clientes de bajo gasto vs clientes de alto gasto) sin que nosotros le digamos de antemano a qué
# grupo pertenece
```

```python
# 3. Aplicar K-Means para agrupar los datos
# n_clusters=2: Queremos agrupar los datos en 2 clústeres.
# random_state=42: Asegura que los resultados sean reproducibles.
modelo_kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

# El método fit entrena el modelo y encuentra los clústeres.
modelo_kmeans.fit(datos_clientes)
```

KMeans(...): Aquí se crea una instancia del modelo, lo que significa que estamos configurando el algoritmo antes de usarlo. Le estamos pasando tres hiperparámetros:

n_clusters=2: Este es el hiperparámetro más importante. Le estamos diciendo al algoritmo que queremos que encuentre y forme exactamente 2 grupos (clústeres) en los datos. El número de clústeres (K) es un valor que debemos decidir de antemano.

random_state=42: Este parámetro asegura que, cada vez que ejecutes el código, los puntos de inicio de los clústeres sean los mismos. Esto es crucial para la reproducibilidad, ya que el algoritmo de K-Means comienza con una selección aleatoria de los centros de los clústeres.

n_init=10: Este hiperparámetro especifica el número de veces que el algoritmo K-Means se ejecutará con diferentes puntos de inicio aleatorios. El modelo elegirá el mejor resultado de todas estas ejecuciones. Un valor más alto reduce la probabilidad de encontrar un resultado subóptimo. (Nota: en versiones más recientes de scikit-learn, este parámetro por defecto ha sido cambiado para una inicialización más robusta, pero es bueno saber qué significa).

modelo_kmeans.fit(datos_clientes)
fit(...): Este es el método que entrena el modelo. Le estamos pasando nuestro conjunto de datos (datos_clientes) para que el algoritmo lo analice.

¿Qué hace fit?:

Inicialización: Selecciona aleatoriamente 2 puntos de datos (ya que n_clusters=2) como los centros iniciales de los clústeres.

Asignación: Asigna cada punto de datos al centro de clúster más cercano.

Actualización: Recalcula la posición de los centros de los clústeres basándose en el promedio de todos los puntos asignados a ese clúster.

Iteración: Repite los pasos 2 y 3 hasta que los centros de los clústeres ya no se muevan significativamente o hasta que se alcance el número máximo de iteraciones.

Al final de este proceso, el modelo modelo_kmeans ha "aprendido" la estructura de los datos y ha definido los 2 clústeres. El resultado de este entrenamiento se guarda internamente en el objeto modelo_kmeans para que podamos usarlo más adelante (por ejemplo, para ver a qué clúster se asignó cada cliente con modelo_kmeans.labels_).

```python
# 4. Obtener las etiquetas de clúster asignadas a cada cliente
# modelo.labels_ contiene el clúster (0 o 1) asignado a cada fila de datos.
etiquetas_cluster = modelo_kmeans.labels_
print("Etiquetas de clúster asignadas a los clientes:", etiquetas_cluster)
```

modelo_kmeans.labels_: Después de que el método fit() termina de ejecutarse, el objeto modelo_kmeans guarda el resultado de la clusterización. El atributo .labels_ es un array de NumPy que contiene las etiquetas de clúster asignadas a cada punto de datos.

La primera etiqueta corresponde al primer cliente en el array de datos.

La segunda etiqueta corresponde al segundo cliente, y así sucesivamente.

Las etiquetas son números enteros (en este caso, 0 o 1 porque le dijimos al modelo que buscara 2 clústeres).

etiquetas_cluster = ...: Esta línea simplemente guarda ese array de etiquetas en una nueva variable llamada etiquetas_cluster para que sea más fácil de usar más tarde.

print(...): Esta línea de código imprime en la consola las etiquetas que el modelo ha asignado. Esto te permite ver de forma textual a qué grupo ha sido clasificado cada uno de los clientes.

¿Qué hace esta parte?
Esta sección del código toma el resultado del entrenamiento del modelo K-Means y lo hace accesible para que lo puedas inspeccionar. Sin esta parte, no podríamos saber a qué clúster ha asignado el algoritmo a cada cliente. La visualización de estas etiquetas (como se haría en un gráfico) es lo que te permite entender la estructura de los grupos que el algoritmo ha descubierto.

```python
# 5. Visualizar los resultados de la clusterización
# Graficamos los clientes y los coloreamos según el clúster asignado.
plt.figure(figsize=(8, 6))
plt.scatter(
    datos_clientes[:, 0], datos_clientes[:, 1],
    c=etiquetas_cluster,  # El color de cada punto depende de su etiqueta de clúster.
    cmap='viridis',       # 'viridis' es la paleta de colores.
    s=100                 # Tamaño de los puntos.
)
plt.xlabel("Ingreso")
plt.ylabel("Gasto Promedio Mensual")
plt.title("Clusterización de Clientes con K-Means")
plt.grid(True)
plt.show()
```

plt.figure(figsize=(8, 6)): Crea una "figura" o ventana de gráfico con un tamaño de 8 pulgadas de ancho por 6 pulgadas de alto.

plt.scatter(...): Esta es la función principal que crea el gráfico de dispersión (un gráfico de puntos). Aquí se le pasan los argumentos clave:

datos_clientes[:, 0]: Le dice a Matplotlib que use la primera columna de nuestro array de datos (datos_clientes) para el eje X. En nuestro ejemplo, esta columna representa el ingreso del cliente.

datos_clientes[:, 1]: Le dice a Matplotlib que use la segunda columna para el eje Y. En este caso, representa el gasto promedio mensual.

c=etiquetas_cluster: Este es el argumento más importante para la visualización de la clusterización. Le dice a la función scatter que el color de cada punto debe basarse en el valor de la etiqueta de clúster que le asignamos. Los puntos del clúster 0 tendrán un color, y los del clúster 1 tendrán otro.

cmap='viridis': Especifica la paleta de colores que se usará para diferenciar los clústeres.

s=100: Define el tamaño de cada punto en el gráfico.

plt.xlabel(...), plt.ylabel(...), plt.title(...): Estas líneas añaden etiquetas a los ejes X e Y, y un título al gráfico, haciendo que la visualización sea clara y fácil de entender.

plt.grid(True): Añade una cuadrícula al gráfico para una mejor lectura de los valores.

plt.show(): Muestra la figura con el gráfico.

Esta parte del código convierte los resultados numéricos del algoritmo K-Means (etiquetas_cluster) en una representación visual, lo que permite a tus estudiantes ver directamente cómo el algoritmo ha segmentado a los clientes en dos grupos distintos. Un grupo tendrá un color y el otro grupo tendrá un color diferente, y la separación en el gráfico mostrará por qué el algoritmo los agrupó de esa manera.

Interpretación del Gráfico

Eje X (Ingreso): Representa el ingreso de cada cliente.


Eje Y (Gasto Promedio Mensual): Representa el gasto promedio mensual de cada cliente.


Los Puntos: Cada punto en el gráfico representa a un cliente individual.


Los Colores: Los colores (azul y amarillo/verde) representan los clústeres o grupos que el algoritmo K-Means ha descubierto. El código le dijo a K-Means que dividiera los datos en dos clústeres (

n_clusters=2), por lo que vemos dos colores distintos.

Análisis de los Clústeres
El gráfico muestra claramente que el algoritmo ha agrupado a los clientes en dos segmentos distintos, basándose en la similitud de sus ingresos y gastos:

```python
# Ejemplo 2: Reducción de Dimensionalidad con PCA
# Este código demuestra cómo utilizar el algoritmo

# Principal Component Analysis (PCA) para reducir la dimensionalidad de un conjunto de datos. Es ideal para simplificar grandes volúmenes
# de datos y hacerlos más manejables para visualización o para otros modelos de machine learning.
```

```python
# ==============================================================================
# SECCIÓN: REDUCCIÓN DE DIMENSIONALIDAD CON PCA
# Este código demuestra cómo usar el algoritmo PCA para reducir el número de
# variables de un dataset, conservando la información más relevante.
# ==============================================================================

# 1. Importar las librerías necesarias
# PCA: El algoritmo de Principal Component Analysis.
# numpy: Para la manipulación de arrays y la generación de datos.
from sklearn.decomposition import PCA
import numpy as np

print("--- DEMOSTRACIÓN DE REDUCCIÓN DE DIMENSIONALIDAD CON PCA ---")
```

```python
# 2. Generar un dataset de ejemplo con 5 características (variables)
# datos.shape será (100, 5), es decir, 100 muestras con 5 variables.
np.random.seed(42)
datos_originales = np.random.rand(100, 5)

print("Forma original de los datos:", datos_originales.shape)

# datos_originales = np.random.rand(100, 5): Aquí se crea un array de NumPy que sirve como nuestro conjunto de datos de ejemplo.

# np.random.rand(...) genera un array lleno de números aleatorios.

# El argumento (100, 5) define la forma o las dimensiones de ese array. Esto significa que el conjunto de datos tendrá 100 filas (muestras) y
# 5 columnas (características o variables).

# print("Forma original de los datos:", datos_originales.shape): Esta línea simplemente imprime en la consola la forma del array que acabamos de
# crear, que será (100, 5). Esto confirma visualmente que el dataset tiene 100 muestras y 5 variables antes de aplicar cualquier reducción de dimensionalidad.
```

```python
# La Reducción de Dimensionalidad es una técnica de aprendizaje no supervisado que busca simplificar un conjunto de datos grande eliminando
# variables redundantes, pero conservando la mayor cantidad de información posible. Esto es muy útil para tareas como el preprocesamiento de
# datos para otros modelos, la visualización de datos complejos y la identificación de las características más relevantes.
# PCA (Principal Component Analysis) es uno de los métodos más populares para lograr esta reducción. Lo que hace es transformar el conjunto de datos en un nuevo sistema de coordenadas donde las nuevas variables (llamadas "componentes principales") capturan la mayor varianza de los datos originales.

# 3. Aplicar PCA para reducir la dimensionalidad
# n_components=2: Queremos reducir las 5 variables originales a solo 2.
pca = PCA(n_components=2)

# El método fit_transform() entrena el modelo PCA y aplica la transformación
# a los datos en un solo paso.
datos_reducidos = pca.fit_transform(datos_originales)
```

```python
# 4. Mostrar la forma (shape) de los datos antes y después de la reducción
print("Forma de los datos reducidos:", datos_reducidos.shape)
```

plt.figure(figsize=(8, 6)): Crea una figura o lienzo para el gráfico con un tamaño de 8 pulgadas de ancho por 6 pulgadas de alto.

plt.scatter(...): Esta es la función clave que genera el gráfico de dispersión (un gráfico de puntos). Los argumentos son:

datos_reducidos[:, 0]: Le indica a Matplotlib que use la primera columna del array datos_reducidos para el eje X. Esta columna corresponde a la primera componente principal que PCA calculó.

datos_reducidos[:, 1]: Le indica que use la segunda columna para el eje Y. Esta es la segunda componente principal.

alpha=0.7: Este parámetro ajusta la transparencia de los puntos. Un valor de 0.7 hace que los puntos sean semitransparentes, lo que ayuda a visualizar la densidad de los datos si hay muchos puntos superpuestos.

plt.xlabel(...) y plt.ylabel(...): Añaden etiquetas descriptivas a los ejes X e Y. Es importante llamarlos "Componente Principal 1" y "Componente Principal 2" para reflejar que ya no estamos trabajando con las variables originales.

plt.title(...): Asigna un título al gráfico para que sea claro qué estamos visualizando.

plt.grid(True): Añade una cuadrícula al gráfico para facilitar la lectura.

plt.show(): Muestra el gráfico en pantalla.

En resumen, esta parte del código es la que hace tangible el resultado del PCA. Nos permite ver la estructura y la distribución de los datos en un espacio bidimensional, lo cual es muy útil para identificar patrones o clústeres que podrían no haber sido evidentes en las 5 dimensiones originales.

```python
# 5. (Opcional) Visualizar los datos reducidos
# Ahora que tenemos solo 2 dimensiones, podemos graficarlos.
plt.figure(figsize=(8, 6))
plt.scatter(datos_reducidos[:, 0], datos_reducidos[:, 1], alpha=0.7)
plt.xlabel("Componente Principal 1")
plt.ylabel("Componente Principal 2")
plt.title("Datos Originales Reducidos a 2 Dimensiones con PCA")
plt.grid(True)
plt.show()
```

Título: El título "Datos Originales Reducidos a 2 Dimensiones con PCA" nos dice claramente qué estamos viendo: los datos que originalmente tenían 5 variables (features) ahora se han transformado en solo 2.

Eje X ("Componente Principal 1"): Esta es la nueva variable que PCA ha creado. Representa la dirección en los datos originales donde hay la mayor varianza. Es la dirección que captura la mayor cantidad de información de los datos.

Eje Y ("Componente Principal 2"): Esta es la segunda nueva variable. Representa la dirección con la segunda mayor varianza, y es ortogonal (perpendicular) a la primera componente principal.

Los Puntos Azules: Cada punto en el gráfico representa una de las 100 muestras originales. La posición de cada punto en este plano 2D es el resultado de la transformación de sus 5 variables originales a sus nuevas coordenadas en las dos componentes principales.

Conclusión
El gráfico muestra la estructura de los datos originales en un formato que es fácil de visualizar. Aunque los datos originales tenían 5 dimensiones, la reducción a 2 nos permite ver si hay patrones, como clústeres o una distribución particular.

En este caso, el gráfico muestra una distribución de puntos dispersa y sin un patrón claro, lo que era de esperar, ya que los datos de ejemplo fueron generados aleatoriamente (np.random.rand). Si los datos originales hubieran tenido una estructura subyacente (por ejemplo, si estuvieran agrupados en clústeres), veríamos esas agrupaciones visualmente en este gráfico.

```python
# 1. CLUSTERIZACIÓN: K-means con un dataset simple
# Importar librerías necesarias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
# KMeans: el algoritmo de clustering no supervisado que agrupa datos en K grupos (clústeres).

# make_blobs: genera automáticamente datos simulados para probar algoritmos de clustering.

# Generar datos artificiales (para simular 3 clústeres)
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=42)
# Esta línea genera un conjunto de datos con estas características:

# Parámetro	Significado
# n_samples=300	Se generarán 300 puntos de datos (observaciones).
# centers=3	Se generarán 3 grupos o clústeres verdaderos.
# cluster_std=0.60	El grado de dispersión (desviación estándar) de los puntos dentro de cada clúster.
# random_state=42	Fija la semilla aleatoria para que siempre se generen los mismos datos.

# X: es una matriz de tamaño (300, 2) que contiene las coordenadas x e y de los puntos generados (es decir, datos en 2 dimensiones).
# y_true: son las etiquetas reales (0, 1 o 2) que indican a qué clúster pertenece cada punto, aunque estas etiquetas no se usan por
# K-means, ya que este algoritmo es no supervisado (no necesita saber las clases verdaderas).

# ¿Por qué usamos make_blobs?
# Porque es muy útil para probar visualmente cómo funciona un algoritmo de agrupamiento, ya que genera datos que se agrupan de forma
# natural en distintos clústeres.

# Crear el modelo K-Means con 3 clústeres
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
y_kmeans = kmeans.predict(X)
# KMeans(...): estamos creando una instancia del modelo KMeans.
# n_clusters=3: le decimos al modelo que queremos encontrar 3 grupos o clústeres.
# random_state=42: fija la aleatoriedad para que los resultados sean reproducibles.

# kmeans.fit(X)
# fit(X): entrena el modelo K-Means usando los datos de entrada X.
# Esto significa que el algoritmo:
# Calcula los centroides iniciales (aleatoriamente).
# Asigna cada punto al centro más cercano.
# Ajusta los centroides (promedio de los puntos en cada grupo).
# Repite hasta que los centroides ya no cambien significativamente.

# y_kmeans = kmeans.predict(X)
# predict(X): etiqueta cada punto de X con el clúster al que pertenece.
# El resultado y_kmeans es un arreglo de etiquetas (0, 1 o 2), indicando el grupo de cada punto.



# Graficar resultados
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, c='red', marker='X', label='Centroides')
plt.title("Clusterización con K-Means")
plt.legend()
plt.show()
```

Se ve un gráfico con los 300 puntos agrupados en 3 colores diferentes, y 3 centroides rojos marcados con una “X”. Es una forma visual de ver cómo K-Means agrupa datos no etiquetados.

Objetivo del gráfico:
Visualizar cómo el algoritmo K-Means ha agrupado los datos en 3 clústeres y dónde ubicó los centroides (los puntos centrales de cada grupo).

Puntos de colores (datos)
Cada punto representa una observación o dato con dos características (coordenadas X y Y).

Los puntos están coloreados según el clúster al que fueron asignados por el modelo K-Means:

Un color para cada grupo (por ejemplo, verde, morado, amarillo).

Estos colores indican que los puntos dentro del mismo grupo son similares entre sí y diferentes de los otros grupos.

Centroides (cruces rojas grandes con forma de X)
Son los centros geométricos de cada clúster, es decir, el promedio de las coordenadas de todos los puntos de ese grupo.

El algoritmo K-Means calcula y ajusta estos centroides hasta que no cambian mucho entre iteraciones.

En este gráfico:

Hay tres centroides (porque usamos n_clusters=3).

Están marcados con cruces grandes en color rojo.

El texto “Centroides” aparece en la leyenda del gráfico.

Interpretación general
El algoritmo ha dividido el conjunto de datos en 3 regiones distintas y ha encontrado un punto central (centroide) para cada una.

Por ejemplo:

El grupo inferior izquierdo contiene muchos puntos cercanos, y el centroide rojo está justo en el medio de ellos.

Lo mismo ocurre con los otros dos grupos: uno arriba y uno a la derecha.

```python
# 3. DETECCIÓN DE ANOMALÍAS: usando Isolation Forest
# Objetivo del código:
# Detectar anomalías (outliers) en un conjunto de datos bidimensional usando el algoritmo Isolation Forest.
from sklearn.ensemble import IsolationForest
# Se importa el algoritmo Isolation Forest, que pertenece a los modelos de conjunto (ensemble models). Es muy eficiente para detectar valores atípicos.

# Crear datos artificiales con algunas anomalías
rng = np.random.RandomState(42)
X_normal = rng.normal(0, 0.5, size=(100, 2))
X_outliers = rng.uniform(low=-4, high=4, size=(10, 2))
X_total = np.vstack([X_normal, X_outliers])

# rng = np.random.RandomState(42)
# X_normal = rng.normal(0, 0.5, size=(100, 2))       # Datos normales (centro en 0, desviación 0.5)
# X_outliers = rng.uniform(low=-4, high=4, size=(10, 2))  # Datos anómalos, más dispersos
# X_total = np.vstack([X_normal, X_outliers])        # Juntar todo en una sola matriz
# X_normal: genera 100 puntos con distribución normal (concentrados cerca del centro).

# X_outliers: genera 10 puntos con distribución uniforme amplia (mucho más dispersos).

#X_total: se combinan los datos normales con los atípicos para entrenar el modelo.

# Modelo de detección de anomalías
modelo_iso = IsolationForest(contamination=0.1, random_state=42)
modelo_iso.fit(X_total)
preds = modelo_iso.predict(X_total)
# contamination=0.1: le dice al modelo que esperamos que el 10% de los datos sean anomalías (en este caso, 10 de 110 puntos).
# fit(X_total): el modelo "aprende" qué datos parecen normales y cuáles son raros.
# predict(X_total): hace la predicción. Devuelve:
# 1 si el punto es considerado normal.
# -1 si el punto es considerado anómalo.

# Graficar: -1 = anomalía, 1 = normal
plt.scatter(X_total[:, 0], X_total[:, 1], c=preds, cmap='coolwarm', edgecolors='k')
plt.title("Detección de Anomalías con Isolation Forest")
plt.show()
# Se grafica cada punto según su predicción:
# Colores distintos (c=preds) para normales y anómalos.
# coolwarm es una paleta que resalta bien los contrastes.
# Puedes ver cómo se separan visualmente los valores atípicos del resto.
```

Este gráfico muestra un conjunto de datos bidimensionales en el plano (eje X y eje Y), que fue analizado con el algoritmo Isolation Forest para detectar anomalías (outliers).

Están agrupados en el centro del gráfico, formando una nube densa. Esto tiene sentido porque la mayoría de los datos (100 puntos) fueron generados con una distribución normal centrada en 0.

Puntos de color azul:

Son los datos detectados como anomalías (preds = -1).

Están aislados, lejos del grupo central. Fueron generados con una distribución uniforme entre -4 y 4, por lo que están más dispersos.

Son los 10 puntos que el modelo considera fuera de lo común.

¿Qué es exactamente una anomalía?
Una anomalía (o outlier) es un punto de datos que difiere significativamente del resto del conjunto de datos. En este gráfico, son puntos que están ubicados lejos del grupo principal de observaciones.

¿Cómo lo detecta Isolation Forest?
Este algoritmo "aísla" observaciones aleatoriamente mediante particiones.

Las anomalías son más fáciles de aislar porque están solas o en zonas poco densas.

El algoritmo necesita menos cortes para aislar un outlier, lo que lo hace detectable.

En este caso, se le indicó que aproximadamente el 10% de los datos eran esperados como anomalías (contamination=0.1), y eso es lo que intentó encontrar.

```python
# 4.Clusterización jerárquica con scipy
from scipy.cluster.hierarchy import dendrogram, linkage
# Importa dos funciones de scipy.cluster.hierarchy:

# linkage: calcula las distancias y agrupa los datos jerárquicamente.

# dendrogram: dibuja el dendrograma (diagrama de árbol)

# Crear jerarquía de clústeres
linked = linkage(X, method='ward')
# linked = linkage(X, method='ward')
# Aquí se crea el modelo de agrupamiento jerárquico.

# X es tu conjunto de datos (por ejemplo, los que generaste con make_blobs).
# method='ward' indica que se usará el método de Ward, que:
# Minimiza la varianza dentro de cada grupo.
# Busca fusionar los clústeres que generen el menor incremento en la varianza total.
# Resultado: linked es una matriz que representa las uniones entre clústeres: qué observaciones o grupos se unieron, a qué distancia,
# y cuántos puntos contenían en cada fusión.

# Dendrograma
plt.figure(figsize=(10, 5))
dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=False)
# Genera el dendrograma:
# linked: la información jerárquica que se usó para construirlo.
# orientation='top': dibuja el dendrograma con las hojas (observaciones) en la parte inferior.
# distance_sort='descending': ordena las fusiones por distancia (más grandes primero).
# show_leaf_counts=False: no muestra la cantidad de observaciones en cada hoja (puedes poner True si quieres mostrarla).
plt.title("Dendrograma - Clusterización Jerárquica")
plt.xlabel("Observaciones")
plt.ylabel("Distancia")
plt.show()
```

¿Qué es un dendrograma?
Un dendrograma es una forma de visualizar cómo se fusionan clústeres paso a paso. Se lee de abajo hacia arriba:

Cada línea horizontal representa una fusión.

La altura de la línea indica la distancia (o disimilitud) entre los clústeres que se están fusionando.

Puedes "cortar" el dendrograma a cierta altura para obtener un número específico de clústeres.

¿Cómo puedes explicar esto?
Es una técnica no supervisada de agrupamiento.
No requiere indicar el número de clústeres al inicio.
Es útil para ver la estructura jerárquica de los datos.
Puedes determinar el número óptimo de clústeres cortando el dendrograma en diferentes niveles.

Eje X (horizontal): representa las observaciones individuales (en este caso, puntos de datos del dataset). Están etiquetadas (aunque aquí se ven sobrepuestas por la cantidad).

Eje Y (vertical): representa la distancia o disimilitud entre los grupos que se van fusionando.

Ramas del dendrograma:
Cada unión entre líneas horizontales indica una fusión entre dos clústeres o puntos.

Cuanto más alta es la fusión (es decir, más arriba en el eje Y), más disímiles son los clústeres que se están uniendo.

Las líneas verticales indican qué tan lejos estaban los grupos antes de fusionarse.

¿Cómo saber cuántos clústeres hay?
Hay una forma práctica de determinar el número óptimo de clústeres:

Imagina que haces un corte horizontal en el dendrograma, por ejemplo, a una altura de 12.

Cuenta cuántas líneas verticales se cortan: ese es el número de clústeres.

En este gráfico, si cortas aproximadamente a la altura Y = 10–15, ves que se cruzan 3 grandes ramas, por lo tanto, el número óptimo de clústeres sería 3, lo cual es consistente si usaste make_blobs con centers=3.

En resumen:
Cada punto comienza como un clúster propio.

Los clústeres se fusionan sucesivamente, basándose en la distancia entre ellos.

El dendrograma muestra ese proceso de fusión de forma visual.

Cuanto más alto es el punto de unión, más distintos eran los clústeres que se fusionaron.

Puedes elegir cuántos clústeres quieres observar cortando el dendrograma en un nivel deseado.

**Introducción al análisis:** Segmentación de usuarios de streaming musical
La presente actividad utiliza un conjunto de datos simulado que representa el comportamiento de usuarios en una plataforma de streaming musical. Cada fila representa a un usuario, y las variables observadas son:

```python
# Paso 1 y 2: Crear datos simulados y normalizar

import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.preprocessing import StandardScaler

# Crear datos artificiales para simular comportamiento de usuarios
np.random.seed(42)
df = pd.DataFrame({
    'reproducciones': np.random.poisson(30, 100),
    'duracion_promedio': np.random.normal(120, 15, 100),
    'saltos': np.random.poisson(5, 100),
    'artistas_distintos': np.random.randint(1, 20, 100),
    'likes': np.random.randint(0, 100, 100)
})
print(df)

# Normalización de los datos
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# Guardamos como DataFrame escalado
df_scaled = pd.DataFrame(X_scaled, columns=df.columns)
```

```python
# Paso 3: K-Means para Clusterización
from sklearn.cluster import KMeans

# Aplicamos KMeans con k=3
kmeans = KMeans(n_clusters=3, random_state=42)
df_scaled['cluster'] = kmeans.fit_predict(X_scaled)

# Visualizamos con pares de variables
sns.pairplot(df_scaled, hue='cluster', palette='Set2')
plt.suptitle("Visualización por Clusters", y=1.02)
plt.show()
```

¿Qué puedes interpretar?

Correlación entre variables:

```python
# Paso 4: Reducción de Dimensiones con PCA
from sklearn.decomposition import PCA

# Aplicamos PCA para reducir a 2 dimensiones
pca = PCA(n_components=2)
pca_components = pca.fit_transform(X_scaled)

# Creamos nuevo DataFrame para graficar
df_pca = pd.DataFrame(pca_components, columns=['PC1', 'PC2'])
df_pca['cluster'] = df_scaled['cluster']

# Gráfico 2D por componentes principales
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_pca, x='PC1', y='PC2', hue='cluster', palette='Set2')
plt.title("Clusters proyectados en PCA (2D)")
plt.show()

# Opcional: mostrar varianza explicada por cada componente
print("Varianza explicada por PCA:", pca.explained_variance_ratio_)
```

```python
# Paso 5: Detección de Anomalías
from scipy.spatial.distance import cdist

# Obtenemos los centroides desde el modelo KMeans
centroides = kmeans.cluster_centers_

# Calculamos distancias al centroide correspondiente de cada punto
distancias = []
for i in range(len(X_scaled)):
    cluster_id = df_scaled.loc[i, 'cluster']
    centroide = centroides[cluster_id]
    distancia = np.linalg.norm(X_scaled[i] - centroide)
    distancias.append(distancia)

df_scaled['distancia_centroide'] = distancias

# Identificamos los 5 más alejados (potenciales anomalías)
umbral = df_scaled['distancia_centroide'].nlargest(5).min()
df_scaled['anomalía'] = df_scaled['distancia_centroide'] >= umbral

# Visualizamos anomalías en el gráfico PCA
df_pca['anomalía'] = df_scaled['anomalía']

plt.figure(figsize=(8,6))
sns.scatterplot(data=df_pca, x='PC1', y='PC2', hue='cluster', style='anomalía', palette='Set2')
plt.title("Clusters y Anomalías Detectadas")
plt.show()
```

```python
# Ejemplo: Cálculo de distancias al centroide (K-Means)
# 1. Librerías necesarias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import euclidean

# 2. Dataset simple
data = {
    'x': [1, 2, 1, 8, 9, 8],
    'y': [2, 1, 1, 8, 9, 9]
}
df = pd.DataFrame(data)

# 3. Escalar datos para mejorar clusterización
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# 4. Aplicar K-Means con k=2
kmeans = KMeans(n_clusters=2, random_state=42)
df['cluster'] = kmeans.fit_predict(X_scaled)

# 5. Obtener los centroides escalados
centroids = kmeans.cluster_centers_

# 6. Calcular la distancia euclidiana de cada punto a su centroide
distancias = []
for i, row in df.iterrows():
    punto = X_scaled[i]
    centroide = centroids[int(row['cluster'])]
    distancia = euclidean(punto, centroide)
    distancias.append(distancia)

df['distancia_al_centroide'] = distancias

# 7. Mostrar resultados
print(df)

# 8. Visualizar
plt.figure(figsize=(6, 6))
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=df['cluster'], cmap='Set1', s=100)
plt.scatter(centroids[:, 0], centroids[:, 1], c='black', s=200, marker='X', label='Centroides')
plt.title("Clusters con centroides y distancias")
plt.legend()
plt.grid(True)
plt.show()
```

2. Columna cluster

Hay dos clústeres: 0 y 1.

Puntos con cluster = 0 son similares entre sí (grupo más cercano al origen).

Puntos con cluster = 1 forman otro grupo (más alejados del origen).

3. Columna distancia_al_centroide
Es la distancia euclidiana (en el espacio estandarizado) entre el punto y el centroide (centro del clúster) al que fue asignado.

Una distancia más baja significa que el punto está más cerca del centroide, es decir, más representativo del grupo.

```python
# Más ejmplos
# 1. Clusterización con K-Means
# Este código agrupa clientes según edad y gasto. El modelo no sabe qué buscar, pero encuentra patrones por sí solo.
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt

# Datos simulados: [edad, gasto mensual]
datos = np.array([[30, 300], [40, 400], [25, 250], [35, 350], [50, 500]])

# Crear modelo K-Means con 2 grupos
modelo = KMeans(n_clusters=2, random_state=42)
modelo.fit(datos)

# Mostrar etiquetas asignadas a cada punto
print("Etiquetas de cluster:", modelo.labels_)

# Visualizar los clusters
plt.scatter(datos[:, 0], datos[:, 1], c=modelo.labels_, cmap='viridis')
plt.title("Clusterización de clientes")
plt.xlabel("Edad")
plt.ylabel("Gasto mensual")
plt.show()
```

```python
# 3. Detección de Anomalías con Gaussian Mixture
# Explicación para estudiantes: Este modelo detecta puntos que no encajan con el patrón general. Útil para encontrar fraudes o errores.
from sklearn.mixture import GaussianMixture
import numpy as np

# Datos simulados: mayoría cerca de [50, 50], uno fuera de rango
datos = np.array([[50, 50], [52, 48], [49, 51], [100, 100]])

# Modelo de mezcla gaussiana con 1 componente
modelo = GaussianMixture(n_components=1, random_state=42)
modelo.fit(datos)

# Calcular probabilidad de pertenencia
scores = modelo.score_samples(datos)
# Esta función devuelve el logaritmo de la probabilidad de cada punto bajo el modelo gaussiano. Es decir:

# Valores menos negativos (más cercanos a 0) → el punto encaja bien en el patrón.

# Valores más negativos → el punto es menos probable, podría ser una anomalía.
print("Probabilidades logarítmicas:", scores)
# Probabilidades logarítmicas: [-5.89, -7.04, -6.61, -7.18]

# Detectar anomalías (valores muy bajos)
umbral = np.percentile(scores, 25)
anomalías = datos[scores < umbral]
print("Datos anómalos detectados:", anomalías)
#
# Índice	Punto	Score (log probabilidad)	Interpretación
# 0	[50, 50]	-5.89	Muy probable
# 1	[52, 48]	-7.04	Probable
# 2	[49, 51]	-6.61	Probable
# 3	[100, 100]	-7.18	Menos probable → ⚠️
# Datos anómalos detectados: [[100 100]]
# umbral = np.percentile(scores, 25)
# anomalías = datos[scores < umbral]
# Calcula el percentil 25 como umbral.

# Detecta los puntos con probabilidad logarítmica más baja (más negativos).

# En este caso, el punto [100, 100] fue el único que cayó por debajo del umbral → se considera anómalo.
# El modelo aprende cómo se ven los datos normales. Luego calcula qué tan probable es cada punto. Si un punto tiene una probabilidad
# muy baja, lo marcamos como anómalo. En este ejemplo, el cliente con [100, 100] no se parece a los demás, así que lo detectamos como
# fuera de lo común.”
```
